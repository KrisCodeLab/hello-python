## 26 - Datenanalyse mit Pandas

Die Bibliothek `pandas` baut auf NumPy auf und ist das Standardwerkzeug, um tabellarische Daten (wie CSV- oder Excel-Dateien) einzulesen, zu filtern und zu strukturieren. 

Die zentrale Datenstruktur ist das **DataFrame**. Ein DataFrame ist vergleichbar mit einer Excel-Tabelle: Es besitzt benannte Spalten (die verschiedene Datentypen enthalten können) und nummerierte Zeilen. Konventionsgemäß wird die Bibliothek mit dem Kürzel `pd` importiert.

```python
import pandas as pd

# Eine CSV-Datei als DataFrame einlesen (Trennzeichen kann per 'delimiter' angepasst werden)
df = pd.read_csv("daten.csv", delimiter=";")

# Einen schnellen Überblick über die ersten fünf Zeilen der Tabelle erhalten
print("Erste Zeilen:")
print(df.head())

# Die Gesamtanzahl der Einträge (Zeilen) abfragen
anzahl_zeilen = len(df)
print(f"\nAnzahl der Einträge: {anzahl_zeilen}")
```

### Zugriff auf Spalten und Zeilen
Der Zugriff auf bestimmte Bereiche eines DataFrames erfolgt über spezifische Selektoren:
* **Spaltenzugriff:** Erfolgt über den Namen der Spalte in eckigen Klammern.
* **Zeilenzugriff:** Erfolgt über die Methode `.iloc[]` (Integer Location) anhand der numerischen Zeilenposition. Auch "List Slicing" ist hier möglich.

```python
# Zugriff auf eine komplette Spalte
sensor_spalte = df["Sensor"]
print("Gesamte Spalte 'Sensor' anzeigen (inkl. Spaltenname und Doctype):")
print(sensor_spalte)

# Zugriff auf eine spezifische Zeile (hier die allererste Zeile am Index 0)
erste_zeile = df.iloc[0]
print("\nErste Zeile anzeigen (inkl. Index und Doctype):")
print(erste_zeile)


# Zugriff auf einen spezifischen Eintrag (Spalte "Wert" der ersten Zeile)
eintrag = df.iloc[0]["Wert"]
print("\nSpalte 'Wert' in erster Zeile anzeigen:")
print(eintrag)

# Slicing: Die letzten 3 Zeilen des DataFrames ausgeben
letzte_zeilen = df.iloc[-3:]
print("\nDie letzten drei Einträge anzeigen:")
print(letzte_zeilen)
```

### Daten filtern
DataFrames lassen sich extrem effizient durch logische Vergleiche filtern. Die Syntax ist dabei identisch zum Filtern in NumPy-Arrays. Das Ergebnis ist eine neue, gefilterte Kopie der Daten. Sollen Daten über mehrere Kriterien hinweg gefiltert werden, können diese Schritte nacheinander auf Zwischenvariablen angewendet werden.

```python
# Filter 1: Alle Einträge mit Sensor Bodenfeuchte suchen
df_bodenfeuchte = df[df["Sensor"] == "Bodenfeuchte"]
print("Alle Bodenfeuchte-Werte:")
print(df_bodenfeuchte)

# Filter 2: Aus der ersten Filterung nur noch die Einträge behalten, deren Wert über 63 liegt
df_bodenfeuchte_ab_63 = df_bodenfeuchte[df_bodenfeuchte["Wert"] > 63]
print("\nBodenfeuchte-Werte über 63:")
print(df_bodenfeuchte_ab_63.head())
```

### Daten sortieren und Iteration
Um Datensätze anhand einer bestimmten Spalte zu ordnen, wird die Methode `.sort_values()` verwendet. Standardmäßig wird aufsteigend sortiert.

Soll ein DataFrame programmgesteuert Zeile für Zeile durchlaufen werden (was vermieden werden sollte, falls eine vektorisierte Operation möglich ist), wird die Methode `.iterrows()` genutzt. Diese liefert jeweils die numerische Position und die eigentlichen Zeilendaten zurück.

```python
# Daten absteigend nach der Spalte "Wert" sortieren, um die höchsten Messwerte zu finden
df_sortiert = df.sort_values("Wert", ascending=False)

print("Die Top 2 höchsten Messwerte:")
# Zeilenweises Iterieren durch die ersten drei Ergebnisse des sortierten DataFrames
for position, zeile in df_sortiert.head(2).iterrows():
    # Gibt den Sensornamen und den zugehörigen Wert der aktuellen Zeile aus
    print(f"Index {position} | Sensor: {zeile['Sensor']} | Erfasster Wert: {zeile['Wert']}")
```
