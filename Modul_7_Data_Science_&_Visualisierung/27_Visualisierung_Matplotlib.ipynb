## 27 - Visualisierung mit Matplotlib

Nach der Auswertung von Datenreihen (beispielsweise mit NumPy oder Pandas) ist die grafische Darstellung der nächste logische Schritt. In Python hat sich hierfür die Bibliothek `matplotlib` als Standard etabliert. 

Für die Erstellung von Diagrammen wird spezifisch das Modul `pyplot` importiert. Konventionsgemäß wird hierfür das Kürzel `plt` verwendet. 

### Liniendiagramme erstellen
Die grundlegendste Form der Visualisierung ist das Liniendiagramm, welches durch die Methode `.plot()` erzeugt wird. Dieser Methode werden zwei Listen oder Arrays übergeben: Die erste für die X-Achse, die zweite für die Y-Achse. Um die Grafik abschließend auf dem Bildschirm auszugeben, muss zwingend der Befehl `plt.show()` aufgerufen werden.

```python
import matplotlib.pyplot as plt

# Beispieldaten: Tage und zugehörige Temperaturwerte
tage = [1, 2, 3, 4, 5]
temperatur = [22.5, 24.1, 23.8, 25.5, 26.2]

# Ein einfaches Liniendiagramm zeichnen
plt.plot(tage, temperatur)

# Diagramm anzeigen
plt.show()
```

### Diagramme formatieren und konfigurieren
Diagramme lassen sich vielfältig anpassen. Die Methode `.plot()` akzeptiert weitere Parameter zur optischen Formatierung:
* **`color`:** Bestimmt die Farbe der Linie (z. B. als Hexadezimalwert wie `"#ff0000"` für Rot oder Farbnamen wie `"r"`).
* **`linestyle`:** Verändert die Art der Linie (z. B. `"dashed"` für gestrichelt).
* **`marker`:** Setzt ein Symbol an die exakten Datenpunkte (z. B. `"o"` für Kreise oder `"x"` für Kreuze).
* **`label`:** Vergibt einen Namen für die Datenreihe. Damit dieser Name angezeigt wird, muss zusätzlich `plt.legend()` aufgerufen werden.

Achsenbeschriftungen lassen sich mit `plt.xlabel()` und `plt.ylabel()` hinzufügen.

```python
x_werte = [1, 2, 3, 4]
y_werte = [10, 15, 13, 18]

# Formatierte Linie mit Legenden-Label
plt.plot(x_werte, y_werte, color="#f27126", linestyle="dashed", marker="o", label="Umsatzverlauf")

# Achsenbeschriftungen hinzufügen
plt.xlabel("Quartal")
plt.ylabel("Umsatz in Tsd. Euro")

# Legende aktivieren
plt.legend()

plt.show()
```

### Weitere Diagrammtypen
Neben Liniendiagrammen bietet Matplotlib Funktionen für viele weitere Visualisierungsformen:
* **Balkendiagramme (`.bar()`):** Erfordern X- und Y-Werte. Eignen sich für Vergleiche zwischen Kategorien.
* **Punktediagramme / Scatterplots (`.scatter()`):** Zeigen Datenpunkte ohne verbindende Linien, optimal um Korrelationen oder Streuungen darzustellen.
* **Kreisdiagramme (`.pie()`):** Erhalten eine Liste von Werten und berechnen automatisch die Anteile eines Kuchens.

```python
kategorien = ["Sensor A", "Sensor B", "Sensor C"]
werte = [45, 80, 25]

# 1. Balkendiagramm
plt.bar(kategorien, werte, color="tab:blue")
plt.show()

# 2. Punktediagramm (Scatterplot)
plt.scatter([1, 2, 3, 4, 5], [5, 7, 6, 8, 7], color="red", marker="x")
plt.show()

# 3. Kreisdiagramm
anteile = [30, 50, 20]
plt.pie(anteile, labels=["Bereich 1", "Bereich 2", "Bereich 3"])
plt.show()
```
