## 25 - Numerische Berechnungen mit NumPy

Die externe Bibliothek `NumPy` ist auf numerische Berechnungen großer Datenmengen hin optimiert und bildet das Fundament für fast alle Data-Science-Anwendungen in Python. 

Anstelle von Standard-Listen verwendet NumPy eine eigene Datenstruktur: das **Array**. Arrays verhalten sich ähnlich wie Listen, belegen jedoch deutlich weniger Arbeitsspeicher und ermöglichen "vektorisiertes" Rechnen. Das bedeutet, dass mathematische Operationen auf alle Elemente gleichzeitig angewendet werden, ohne dass eine langsame `for`-Schleife geschrieben werden muss.

```python
import numpy as np

# Ein Array aus einer bestehenden Liste erstellen
daten = np.array([1, 2, 3, 4, 5, 10])
print("Array:", daten)

# Arrays automatisch generieren
zahlenreihe = np.arange(1, 11)
print("Zahlenreihe von 1 bis 10:", zahlenreihe)

# Arrays mit Platzhaltern füllen
nullen = np.zeros(5)
einsen = np.ones(5)
print("Fünf Nullen:", nullen)
print("Fünf Einsen:", einsen)
```

### Performantes Rechnen und Statistik
Wird eine Rechenoperation auf ein Array-Objekt angewendet, führt der Interpreter diese Operation in hochoptimiertem C-Code für jeden einzelnen Wert aus. Zudem liefert NumPy fertige, schnelle Methoden für statistische Auswertungen mit.

```python
# Vektorisiertes Rechnen (ersetzt for-Schleifen oder List-Comprehensions)
basis_werte = np.arange(10)
quadrate = basis_werte ** 2
print("Quadrierte Werte:", quadrate)

messdaten = np.array([1.5, 2.5, 3.0, 4.5, 10.0])

# Integrierte statistische Auswertungen
print("Mittelwert:", messdaten.mean())
print("Maximum:", messdaten.max())
print("Standardabweichung:", messdaten.std()) # Zeigt die Streuung (Abweichung vom Durchschnitt) der einzelnen Werte 
```

### Mehrdimensionale Arrays und `reshape()`
Arrays können nicht nur eindimensional (wie eine simple Liste), sondern auch mehrdimensional (als Matrizen) aufgebaut sein. 

Mit der Methode `.reshape()` lässt sich die Struktur eines Arrays verändern. Dabei wird ein Tupel übergeben, welches die neuen Dimensionen (Zeilen und Spalten) exakt definiert. Wird für eine Dimension der Wert `-1` übergeben, berechnet NumPy die benötigte Größe für diese Ebene automatisch anhand der vorhandenen Elemente.

```python
# Ein eindimensionales Array mit 8 Elementen
a = np.array([1, 2, 3, 4, 5, 6, 7, 8])

# Umstrukturierung in 2 Zeilen und 4 Spalten
matrix = a.reshape((2, 4))
print("2x4 Matrix:\n", matrix)

# Automatische Berechnung der Spaltenanzahl durch den Platzhalter '-1'
# Es werden 4 Zeilen vorgegeben, die Spaltenanzahl (hier 2) wird berechnet
matrix_auto = a.reshape((4, -1))
print("\n4x? Matrix:\n", matrix_auto)

# Abfragen der Dimensionen (Zeilen, Spalten)
print("\nDimensionen der Matrix:", matrix_auto.shape)
```

### Arrays filtern (Boolean Masking)
Das Filtern von Daten in Arrays erfolgt äußerst effizient. Wenn eine Vergleichsoperation auf ein gesamtes Array angewendet wird, entsteht ein neues Array, das nur aus Wahrheitswerten (`True` oder `False`) besteht. Dieses sogenannte "Boolean Array" kann dann als Filtermaske in eckigen Klammern an das ursprüngliche Array übergeben werden. Es bleiben nur die Werte stehen, an deren Position ein `True` resultierte.

```python
messwerte = np.array([1, 2, 3, 4])

# Erstellt eine Maske: [False, False, True, True]
maske = messwerte >= 3
print("Filtermaske:", maske)

# Wendet die Maske an
gefiltert = messwerte[maske]
print("Gefilterte Werte (>=3):", gefiltert)

# Kompakte Schreibweise in einem einzigen Schritt
kurz_gefiltert = messwerte[messwerte < 3]
print("Kompakt gefiltert (<3):", kurz_gefiltert)
```
