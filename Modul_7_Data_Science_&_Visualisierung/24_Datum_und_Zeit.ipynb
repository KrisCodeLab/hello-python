## 24 - Datum und Zeit

Zur Verarbeitung und Darstellung von Datumswerten und Uhrzeiten wird in Python das Standard-Modul `datetime` verwendet. Es stellt verschiedene Klassen zur Verfügung, um Zeiten zu repräsentieren und damit zu rechnen:

* **`date`:** Repräsentiert ausschließlich ein Datum (Jahr, Monat, Tag).
* **`time`:** Repräsentiert ausschließlich eine Uhrzeit (Stunde, Minute, Sekunde).
* **`datetime`:** Kombiniert Datum und Uhrzeit in einem Objekt.

```python
import datetime as dt

# Ein spezifisches Datum erstellen
d = dt.date(2026, 6, 4) # Syntax: Y, M, D
print("Datum:", d)

# Eine spezifische Uhrzeit erstellen
t = dt.time(8, 39, 30)
print("Uhrzeit:", t)

# Das aktuelle Datum inkl. Uhrzeit abfragen
jetzt = dt.datetime.now()
print("Aktueller Zeitstempel:", jetzt)
```

### Rechnen mit Zeitdifferenzen (`timedelta`)
Zeitunterschiede werden durch die Klasse `timedelta` abgebildet. Wird ein Datum von einem anderen subtrahiert, resultiert daraus automatisch ein `timedelta`-Objekt, aus dem sich beispielsweise die verbleibenden Tage oder Sekunden extrahieren lassen.

```python
# Zwei Zeitpunkte definieren
jetzt = dt.datetime.now()
heute_nachmittag = dt.datetime(2026, 9, 13, 17, 0, 0)

# Zeitdifferenz berechnen
verbleibende_zeit = heute_nachmittag - jetzt

print("Typ des Ergebnisses:", type(verbleibende_zeit))
print("Verbleibende Zeit:", verbleibende_zeit)

# Umrechnung der restlichen Sekunden in volle Stunden
rest_stunden = verbleibende_zeit.seconds // 3600
print("Das entspricht vollen Stunden:", rest_stunden)
```

### Formatierung und Parsing (`strftime` und `strptime`)
Für die Ein- und Ausgabe müssen Datumsobjekte oft in Textschreibweisen umgewandelt werden (und umgekehrt).
* **`strftime` (String Format Time):** Wandelt ein `datetime`-Objekt anhand von Formatierungs-Codes (z. B. `%H:%M` für Stunde:Minute) in einen formatierten Text (String) um.
* **`strptime` (String Parse Time):** Liest einen Text ein und übersetzt ihn anhand des angegebenen Formats zurück in ein verwertbares `datetime`-Objekt.

```python
import datetime as dt

# 1. strftime: Von datetime zu String
aktuell = dt.datetime.now()
zeit_als_text = aktuell.strftime("%d.%m.%Y %H:%M:%S")
print("Formatiert als Text:", zeit_als_text)

# 2. strptime: Von String zu datetime
zeitangabe = "08:32"
# Das Format %H:%M teilt dem Parser mit, dass zuerst die Stunde und dann die Minute folgt
geparste_zeit = dt.datetime.strptime(zeitangabe, "%H:%M").time()
print("Zurückgewandelt in ein time-Objekt:", geparste_zeit)
```

### Unix-Timestamps
Ein Unix-Timestamp ist eine kompakte Repräsentation eines Datums als Zahl. Er zählt die vergangenen Sekunden seit dem 01.01.1970.

```python
# Den aktuellen Timestamp abfragen
aktueller_timestamp = aktuell.timestamp()
print("Aktueller Unix-Timestamp:", aktueller_timestamp)

# Einen Timestamp zurück in ein lesbares Datum umwandeln
aus_timestamp = dt.datetime.fromtimestamp(1512043274)
print("Datum aus fixem Timestamp:", aus_timestamp)
```
