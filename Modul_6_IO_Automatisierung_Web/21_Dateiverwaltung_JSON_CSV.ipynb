### Dateizugriff mit dem `with`-Block
Der Zugriff auf das Dateisystem erfolgt standardmäßig über die Funktion `open()`. Um sicherzustellen, dass Dateiverbindungen auch im Fehlerfall korrekt beendet werden, wird der `with`-Block (Kontext-Manager) verwendet. Dieser macht den manuellen Aufruf von `.close()` überflüssig. Temporäre Dateien, die nach der Nutzung wieder aus dem System gelöscht werden sollen, können zudem mit dem Modul `tempfile` erzeugt werden.

```python
import tempfile

# Sicheres Schreiben in eine Textdatei
with open("ausgabedatei.txt", "w", encoding="utf-8") as file:
    file.write("Systemprotokoll gestartet.\n")

# Sicheres Lesen aus einer Textdatei
with open("ausgabedatei.txt", "r", encoding="utf-8") as file:
    inhalt = file.readlines()
    print("Dateiinhalt:", inhalt)

# Nutzung einer temporären Datei, die das Betriebssystem verwaltet
with tempfile.NamedTemporaryFile() as tf:
    print("Temporäre Datei kurzzeitig erstellt unter:", tf.name)
```

### Strukturierte Daten: CSV und JSON
Für den Datenaustausch und die persistente Speicherung werden in der Praxis häufig strukturierte Formate wie CSV (Comma Separated Values) oder JSON (JavaScript Object Notation) verwendet. Python bietet hierfür die Standard-Module `csv` und `json`.

```python
import csv
import json

# --- JSON Verarbeitung ---
# Ein Dictionary in einen JSON-String umwandeln
daten_dict = {"sensor": "Bodentemperatur", "wert": 25.4, "aktiv": True}
json_string = json.dumps(daten_dict)
print("JSON-Format:", json_string)

# --- CSV Verarbeitung ---
# Schreiben einer CSV-Datei (newline="" verhindert leere Zeilen unter Windows)
with open("messwerte.csv", "w", newline="", encoding="utf-8") as file:
    csv_writer = csv.writer(file, delimiter=";")
    csv_writer.writerow(["Sensor", "Wert"])
    csv_writer.writerow(["Bodenfeuchte", 62.5])

# Lesen einer CSV-Datei
with open("messwerte.csv", "r", encoding="utf-8") as file:
    csv_reader = csv.reader(file, delimiter=";")
    for zeile in csv_reader:
        print("Gelesene CSV-Zeile:", zeile)
```
