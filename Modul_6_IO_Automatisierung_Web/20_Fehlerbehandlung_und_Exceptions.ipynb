## 20 - Fehlerbehandlung und Exceptions

In der Softwareentwicklung kommt es unweigerlich zu Laufzeitfehlern (Exceptions), beispielsweise wenn eine nicht existierende Datei geöffnet werden soll oder auf einen falschen Index zugegriffen wird. Ohne Fehlerbehandlung führt eine Exception zum sofortigen Absturz des gesamten Programms. 

Durch einen `try`-`except`-Kontrollfluss lassen sich diese Fehler gezielt abfangen. Der riskante Code wird im `try`-Block ausgeführt. Tritt ein Fehler auf, wird das Programm nicht beendet, sondern springt sofort in den `except`-Block.

```python
daten = [1, 2]

try:
    # Provoziert einen IndexError, da der Index 3 nicht existiert
    print(daten[3])

# Fängt gezielt nur den IndexError ab
except IndexError:
    print("Fehler: Auf diesen Index kann nicht zugegriffen werden.")
    
# Ein allgemeines except fängt alle restlichen Fehler ab
except Exception as e:
    print("Ein unbekannter Fehler ist aufgetreten:", e)
```

### Die Blöcke `else` und `finally`
Die Fehlerbehandlung lässt sich um zwei weitere, nützliche Blöcke erweitern, um den Programmfluss exakt zu steuern:

* **`else`:** Wird nur dann ausgeführt, wenn der `try`-Block zu 100 % fehlerfrei durchlaufen wurde.
* **`finally`:** Wird in jedem Fall ausgeführt – unabhängig davon, ob ein Fehler aufgetreten ist oder nicht. Dieser Block wird in der Praxis genutzt, um Systemressourcen sicher freizugeben (z. B. um Dateizugriffe zu beenden oder Datenbankverbindungen zu schließen).

```python
try:
    filehandle = open("ausgabedatei.txt", "w")
    filehandle.write("Beispieltext")
    
except IOError:
    print("Ausgabe in die Datei fehlgeschlagen.")

else:
    print("Kein Fehler aufgetreten. Der Schreibvorgang war erfolgreich.")

finally:
    # Dieser Block wird garantiert ausgeführt, um die Datei sicher zu schließen
    print("Aufräumarbeiten: Datei wird geschlossen.")
    filehandle.close()
```

### Fehler gezielt auslösen (`raise`) und eigene Fehlerklassen
Manchmal erfordert die Logik einer Anwendung, dass ein Fehler absichtlich ausgelöst wird (beispielsweise bei der Validierung von fehlerhaften Benutzereingaben). In Python wird hierfür das Schlüsselwort `raise` verwendet. 

Es können Standard-Exceptions wie ein `AssertionError` geworfen werden, oder es werden eigene Fehlerklassen definiert, die von der Basisklasse `Exception` erben.

```python
# Beispiel 1: Einen Standard-Fehler auslösen
def level_check(level):
    if level < 1:
        # Bricht das Programm kontrolliert mit einer spezifischen Fehlermeldung ab
        raise AssertionError("Ungültiges Level: Das Level muss mindestens 1 sein.")
    print("Level ist gültig.")

# Testaufruf (wird hier per try/except direkt wieder abgefangen, um einen Absturz zu verhindern)
try:
    level_check(0)
except AssertionError as error:
    print("Abgefangener Fehler:", error)

print("-" * 30)

# Beispiel 2: Eine eigene Fehlerklasse definieren
class EmptyDictionaryError(Exception):
    pass

d = {}

try:
    if len(d) == 0:
        raise EmptyDictionaryError("Das Dictionary enthält kein Element.")
except EmptyDictionaryError as error:
    print("Eigener Fehler abgefangen:", error)
```
