### System-Parameter mit `sys`
Das Modul `sys` stellt Funktionen bereit, um direkt mit dem Python-Interpreter und dem Betriebssystem zu interagieren. Es ermöglicht beispielsweise das Auslesen der verwendeten Python-Version, der Plattform oder das Abfragen und Manipulieren des systemweiten Rekursionslimits.

```python
import sys

# Auslesen von Meta-Informationen zum System und zur Laufzeitumgebung
print("Verwendete Plattform:", sys.platform)
print("Python-Version:", sys.version)

# Das Rekursionslimit bestimmt, wie oft sich eine Funktion selbst aufrufen darf
aktuelles_limit = sys.getrecursionlimit()
print("Aktuelles Rekursionslimit:", aktuelles_limit)
```

### Tastatureingaben automatisieren (`keyboard`)
Externe Module wie `keyboard` ermöglichen es, globale Tastaturereignisse zu registrieren und zu steuern. Damit lassen sich beispielsweise Eingabe-Makros aufzeichnen oder bestimmte Skriptfunktionen durch globale Tastenkombinationen (Hooks) auslösen.

```python
import keyboard

# Zeichnet alle Tastenanschläge auf, bis die Taste 'Esc' gedrückt wird
tasten_events = keyboard.record(until="esc")

# Gibt die aufgezeichneten Anschläge als bereinigten String zurück
for zeichen in keyboard.get_typed_strings(tasten_events, allow_backspace=False):
    print("Aufgezeichnete Eingabe:", zeichen)
```

### Desktop-Automatisierung (`pyautogui`)
Für die automatische Steuerung der grafischen Benutzeroberfläche (Mausbewegungen, Klicks, Tastatureingaben) sowie zur Bildschirmerfassung wird in der Praxis häufig das Modul `pyautogui` eingesetzt. Es erlaubt die vollständige Automatisierung wiederkehrender Desktop-Prozesse.

```python
# Hinweis: Benötigt ebenfalls die Installation über 'pip install pyautogui'
import pyautogui
import os

# Erstellt einen Screenshot des gesamten Bildschirms und speichert ihn als Datei ab.
# Der Speicherpfad wird vorher sichergestellt.

if not os.path.exists("./screenshots"):
    os.mkdir("./screenshots")

speicherort = "./screenshots/test_aufnahme.png"
pyautogui.screenshot(speicherort)
print("Screenshot gespeichert unter:", speicherort)
```
