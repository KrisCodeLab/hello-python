## 22 - Web-Scraping und Netzwerkanfragen

Für die automatisierte Datenbeschaffung aus dem Internet (Web-Scraping) werden in Python typischerweise zwei spezialisierte externe Bibliotheken kombiniert:
* **`requests`:** Sendet HTTP-Anfragen (wie GET oder POST) an einen Webserver und lädt den unstrukturierten Quelltext (HTML) einer Webseite herunter.
* **`BeautifulSoup` (aus dem Paket `bs4`):** Wandelt den rohen HTML-Text in ein durchsuchbares Objekt um. Dadurch lässt sich gezielt nach spezifischen HTML-Tags, CSS-Klassen oder IDs suchen, um die benötigten Informationen zu extrahieren[.

```python
from urllib.parse import urljoin
import requests
from bs4 import BeautifulSoup

# 1. Quelltext der Webseite über eine GET-Anfrage herunterladen
url = "http://python.beispiel.programmierenlernen.io/index.php"
response = requests.get(url)

# 2. Den heruntergeladenen Text als HTML parsen
doc = BeautifulSoup(response.text, "html.parser")

# 3. Gezielt nach Elementen suchen (z. B. nach allen Elementen mit der CSS-Klasse '.card')
karten = doc.select(".card")

if len(karten) > 0:
    erste_karte = karten[0]

    # Text aus einem spezifischen HTML-Element extrahieren
    titel = erste_karte.select(".card-title span")[1].text
    print("Gefundener Titel:", titel)

    # Ein Attribut (hier den Bild-Link 'src') auslesen
    bild_pfad = erste_karte.select_one("img").attrs["src"]

    # Relativen Pfad mit der Basis-URL zu einem gültigen, absoluten Link verknüpfen
    bild_url = urljoin(url, bild_pfad)
    print("Zusammengesetzte Bild-URL:", bild_url)
```

### Ein lokaler HTTP-Server
Python bietet in der Standardbibliothek die Möglichkeit, einen eigenen HTTP-Server zu starten. Das Modul `http.server` kann genutzt werden, um Anfragen (wie GET oder POST) auf einem definierten Port (z. B. 10080) entgegenzunehmen und über eine Handler-Klasse programmatisch darauf zu reagieren.

```python
# Hinweis: Dieser Codeblock konfiguriert einen Server, der potenziell endlos läuft.
# In einem Jupyter Notebook blockiert der Start-Befehl die weitere Ausführung der Zelle,
# weshalb der finale '.serve_forever()'-Befehl hier auskommentiert ist.

from http.server import HTTPServer, BaseHTTPRequestHandler

class EigenerRequestHandler(BaseHTTPRequestHandler):
    # Definiert die Logik, wenn der Server eine GET-Anfrage erhält
    def do_GET(self):
        # 1. HTTP-Statuscode senden (200 = OK)
        self.send_response(200)
        # 2. Header setzen
        self.send_header("Content-Type", "text/plain; charset=utf-8")
        self.end_headers()

        # 3. Antwortnachricht generieren und als Byte-Stream senden
        antwort = "Der Server hat die Anfrage für den Pfad " + self.path + " erhalten."
        self.wfile.write(antwort.encode())

# Server-Konfiguration: Lokalhost (127.0.0.1) auf Port 10080
adresse = ("127.0.0.1", 10080)
server = HTTPServer(adresse, EigenerRequestHandler)

print("Lokaler Server erfolgreich konfiguriert auf http://127.0.0.1:10080")

# Um den Server zu starten, müsste folgende Zeile aktiviert werden:
# server.serve_forever()
```
