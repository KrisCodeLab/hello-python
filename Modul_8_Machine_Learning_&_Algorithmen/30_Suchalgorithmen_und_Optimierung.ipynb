### Lineare vs. Binäre Suche
Bei der Suche nach spezifischen Werten in Datenstrukturen (wie Listen) spielt die Effizienz eine entscheidende Rolle:

* **Lineare Suche:** Das Array oder die Liste wird Element für Element von Anfang bis Ende durchsucht. Die Laufzeit beträgt $O(n)$. Der Vorteil ist, dass die Daten ungeordnet sein dürfen.
* **Binäre Suche:** Teilt den Suchbereich in jedem Schritt in der Mitte und verwirft die Hälften, in denen der gesuchte Wert nicht liegen kann. Die Laufzeit sinkt dadurch drastisch auf $O(\log n)$. **Voraussetzung:** Die Daten müssen zwingend aufsteigend sortiert sein.

```python
# Implementierung einer iterativen binären Suche

def binäre_suche(daten, gesuchter_wert):
    # Voraussetzung: Daten müssen sortiert vorliegen
    daten.sort()

    links = 0
    rechts = len(daten) - 1

    while links <= rechts:
        # Die Mitte des aktuellen Suchbereichs ermitteln (ganzzahlige Division)
        mitte = (links + rechts) // 2

        if daten[mitte] == gesuchter_wert:
            return mitte # Wert gefunden, Index zurückgeben
        elif daten[mitte] < gesuchter_wert:
            links = mitte + 1 # In der rechten Hälfte weitersuchen
        else:
            rechts = mitte - 1 # In der linken Hälfte weitersuchen

    return -1 # Wert ist nicht in der Liste enthalten

# Test der Funktion
zahlen = [45, 12, 67, 3, 89, 105, 46, 40, 33, 20]
ergebnis_index = binäre_suche(zahlen, 46)

print("Index des gesuchten Wertes:", ergebnis_index)
```

### Rekursive binäre Suche
Die binäre Suche lässt sich eleganter auch per Rekursion (Selbstaufruf der Funktion mit verkleinertem Suchbereich) abbilden. Dabei muss jedoch darauf geachtet werden, dass die Abbruchbedingungen (Wert gefunden oder Suchbereich leer) präzise definiert sind.

```python
def binäre_suche_rekursiv(daten, gesuchter_wert, links, rechts):
    if links > rechts:
        return -1

    mitte = (links + rechts) // 2

    if daten[mitte] == gesuchter_wert:
        return mitte
    elif daten[mitte] < gesuchter_wert:
        return binäre_suche_rekursiv(daten, gesuchter_wert, mitte + 1, rechts)
    else:
        return binäre_suche_rekursiv(daten, gesuchter_wert, links, mitte - 1)

sortierte_daten = [3, 12, 20, 33, 40, 45, 46, 67, 89, 105]
index = binäre_suche_rekursiv(sortierte_daten, 89, 0, len(sortierte_daten) - 1)

print("Rekursiv ermittelter Index:", index)
```
