Die Wahl der richtigen Datenstruktur und effizienter Algorithmen entscheidet maßgeblich über die Performance eines Programms.

### 1.) Laufzeitmessung mit `datetime`
Um die Effizienz von Code zu bewerten, kann die Ausführungszeit über Zeitstempel vor und nach der Ausführung gemessen werden.

```python
from datetime import datetime

start = datetime.now()

N = 100
odds = []

for i in range(N):
    if (i % 2 != 0): 
        odds.append(i)
    
print("Alle ungeraden Zahlen bis " + str(N - 1) + ": ", odds)

end = datetime.now()
print("Benötigte Zeit:", end - start)
```

### 2.) Zeitmessung in Jupyter mit `%%timeit`

### 3.) Die O-Notation (Zeitkomplexität)
Die O-Notation beschreibt das Skalierungsverhalten eines Algorithmus in Abhängigkeit von der Datenmenge $n$:
* **$O(1)$ - Konstante Laufzeit:** Die Ausführungszeit bleibt unabhängig von $n$ gleich (z. B. Zugriff auf ein Element per Index oder Prüfung in einem `Set`).
* **$O(n)$ - Lineare Laufzeit:** Die Ausführungszeit steigt proportional zu $n$ (z. B. einfache `for`-Schleife über eine Liste).
* **$O(n^2)$ - Quadratische Laufzeit:** Die Ausführungszeit wächst im Quadrat (z. B. verschachtelte Schleifen).

```python
# Beispiel für O(1) - Suche in einem Set
studenten_set = {"Max", "Erik", "Monika", "Tobias"}
print("O(1) Suche:", "Erik" in studenten_set)

# Beispiel für O(n) - Suche in einer Liste
studenten_liste = ["Max", "Erik", "Monika", "Tobias"]
print("O(n) Suche:", "Erik" in studenten_liste)

# Beispiel für O(n^2) - Quadratische Laufzeit durch verschachtelte Schleifen
def quadratische_laufzeit(n):
    zaehler = 0
    for i in range(n):
        for j in range(n):
            zaehler += 1
    return zaehler

print("Quadratische Laufzeit bei n=10:", quadratische_laufzeit(10))
```
