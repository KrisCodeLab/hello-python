## 31 - Sortieralgorithmen

Um effizient auf Daten zugreifen zu können – wie beispielsweise bei der binären Suche – müssen Datenstrukturen in eine bestimmte Reihenfolge (meist aufsteigend) gebracht werden. In der Informatik gibt es eine Vielzahl von Sortierverfahren mit unterschiedlichen Laufzeiten und Eigenschaften.

### 1. Bubble Sort
Der **Bubble Sort** ist einer der intuitivsten, wenn auch ineffizientesten Sortieralgorithmen ($O(n^2)$). Er vergleicht jeweils zwei benachbarte Elemente und vertauscht sie, falls sie in der falschen Reihenfolge stehen. Größere Elemente "wandern" dadurch schrittweise nach rechts (wie Luftblasen nach oben).

```python
def bubble_sort(daten):
    n = len(daten)
    # Äußere Schleife für die Anzahl der Durchläufe
    for i in range(n):
        # Die letzten i Elemente sind bereits sortiert
        for j in range(0, n - i - 1):
            if daten[j] > daten[j + 1]:
                # Elemente vertauschen (Swapping)
                daten[j], daten[j + 1] = daten[j + 1], daten[j]
    return daten

unsortierte_zahlen = [64, 34, 25, 12, 22, 11, 90]
sortierte_zahlen = bubble_sort(unsortierte_zahlen)

print("Sortierte Liste:", sortierte_zahlen)
```

### 2. Selection Sort
Beim **Selection Sort** wird die Liste in einen bereits sortierten und einen noch unsortierten Bereich unterteilt. Der Algorithmus sucht im unsortierten Bereich kontinuierlich nach dem kleinsten Element und tauscht dieses an den Anfang des unsortierten Bereichs. Die Zeitkomplexität liegt ebenfalls bei $O(n^2)$.

```python
def selection_sort(daten):
    n = len(daten)
    for i in range(n):
        # Annahme: Das aktuelle Element ist das Minimum
        min_index = i
        for j in range(i + 1, n):
            if daten[j] < daten[min_index]:
                min_index = j
        
        # Gefundenes Minimum mit dem ersten unsortierten Element vertauschen
        daten[i], daten[min_index] = daten[min_index], daten[i]
    return daten

unsortierte_werte = [29, 10, 14, 37, 13]
print("Selection Sort Ergebnis:", selection_sort(unsortierte_werte))
```

### 3. Der integrierte Python-Algorithmus: Timsort
In der Praxis implementiert man einfache Sortierverfahren wie Bubble Sort selten selbst, da Python bereits eine hochoptimierte Methode namens **Timsort** (eine Kombination aus Merge Sort und Insertion Sort) bereitstellt. 
* Die Methode `.sort()` sortiert Listen *in-place* (direkt in der bestehenden Liste).
* Die Funktion `sorted()` gibt eine *neue*, sortierte Liste zurück.

```python
namen = ["Max", "Monika", "Erik", "Franziska"]

# 1. In-place sortieren mit .sort()
namen.sort()
print("In-place sortiert:", namen)

# 2. Neue sortierte Liste erzeugen mit sorted()
zahlen = [45, 12, 67, 3, 89]
neue_sortierte_liste = sorted(zahlen, reverse=True)
print("Original unverändert:", zahlen)
print("Neue absteigend sortierte Liste:", neue_sortierte_liste)
```
