### Lineare Regression (`scikit-learn`)
Machine Learning ermöglicht es, Modelle mit historischen Daten zu trainieren, um Muster zu erkennen und Zukunftsprognosen zu berechnen. Ein klassisches Einstiegsbeispiel ist die Lineare Regression, bei der eine Ausgleichsgerade durch vorhandene Datenpunkte (beispielsweise eine Geburtenstatistik im Zeitverlauf) gelegt wird.

Hierfür wird die Bibliothek `scikit-learn` eingesetzt. Vor dem Training (dem sogenannten `fit()`) müssen die Eingabewerte oft umstrukturiert (als höherdimensionale Arrays verpackt) werden.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Exemplarische Datenvorbereitung (fiktive Geburtenraten eines Namens pro Jahr)
jahre = [1950, 1960, 1970, 1980, 1990, 2000]
anzahl = [500, 520, 580, 610, 690, 750]

# Preprocessing: Das Modell erwartet die Eingangsdaten (X) verschachtelt als Liste von Listen
x_train = [[jahr] for jahr in jahre]

# Initialisierung und Training des Modells
model = LinearRegression()
model.fit(x_train, anzahl)

# Eine Zukunftsprognose für das Jahr 2050 berechnen
prognose = model.predict([[2050]])
print("Prognostizierter Wert für 2050:", prognose[0])
```

### Bilderkennung (`OpenCV`)
Die Bibliothek `OpenCV` (importiert als `cv2`) stellt umfangreiche Funktionen für maschinelles Sehen (Machine Vision) bereit. Ein typischer Anwendungsfall ist das Erkennen von Gesichtern oder spezifischen Objekten in Bilddateien. 

Dafür werden vortrainierte Klassifikatoren (wie `CascadeClassifier`) geladen. Um die Verarbeitung zu beschleunigen, werden Bilder vor der Erkennung häufig in Graustufen umgewandelt.

```python
# Hinweis: Die Ausführung dieses Codes setzt eine vorhandene Bilddatei ('bild.jpg') 
# sowie eine XML-Klassifikator-Datei voraus.

import cv2
import matplotlib.pyplot as plt

# 1. Bild einlesen
img = cv2.imread("./resources/bild.jpg")

# 2. Vortrainierten Klassifikator laden
classifier = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_alt.xml")

# 3. Bild in Graustufen umwandeln für effizientere Erkennung
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY) 

# 4. Erkennung durchführen (liefert Koordinaten und Größen der gefundenen Gesichter)
faces = classifier.detectMultiScale(gray, minNeighbors=10)

# 5. Gefundene Bereiche mit einem Rechteck markieren
c = img.copy()
for face in faces:
    x, y, w, h = face
    cv2.rectangle(c, (x, y), (x + w, y + h), (0, 255, 0), 10)

# 6. Farbraum für die korrekte Anzeige in matplotlib anpassen und ausgeben
i = cv2.cvtColor(c, cv2.COLOR_BGR2RGB)
plt.imshow(i)
plt.show()
```
