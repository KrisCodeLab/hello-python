## 13 - Verschachtelungen (Nested Structures)

Datenstrukturen wie Listen und Dictionaries können in Python beliebig tief ineinander verschachtelt werden, um komplexe Datensätze abzubilden. In der Praxis der Softwareentwicklung stellen Verschachtelungen die Standardmethode dar, um relationale Datenstrukturen, JSON-Responses von Web-APIs oder strukturierte CSV-Daten im Arbeitsspeicher abzubilden.

```python
# Listen als Werte in einem Dictionary
fachbereiche = {
    "Informatik": ["Max", "Moritz"],
    "BWL": ["Erik", "Franziska"]
}

# Zugriff auf die gesamte verschachtelte Liste
informatik_studenten = fachbereiche["Informatik"]
print("Informatik:", informatik_studenten)

# Direkter, tiefer Zugriff auf ein spezifisches Element der verschachtelten Liste
erster_bwl_student = fachbereiche["BWL"][0]
print("Erster BWL-Student:", erster_bwl_student)
```

### Dictionaries in Listen (Datensätze)
Ein sehr häufiges Muster ist eine Liste, die mehrere Dictionaries enthält. Jedes Dictionary repräsentiert dabei einen eigenständigen Datensatz mit identischen Schlüsseln (z. B. Zeilen aus einer Datenbank).

```python
# Dictionaries als Elemente einer Liste
sortiment = [
    {"artikel": "Milch", "preis": 1.00},
    {"artikel": "Butter", "preis": 2.50},
    {"artikel": "Eier", "preis": 3.00}
]

# Tiefgreifender Lesezugriff:
# Greift auf das zweite Element der Liste zu und liest dort den Wert des Schlüssels "artikel"
zweiter_artikel = sortiment[1]["artikel"]
print("Zweiter Artikel im Sortiment:", zweiter_artikel)
```

### Tiefes Iterieren und Aggregation
Um verschachtelte Strukturen zu verarbeiten, werden oft Schleifen genutzt, die gezielt auf die inneren Schlüssel oder Indizes zugreifen. Dies wird häufig zur Aggregation von Daten verwendet, beispielsweise um Gesamtwerte oder Durchschnitte zu berechnen.

```python
# Konstruktion eines fiktiven Warenkorbs aus dem bestehenden Sortiment
warenkorb = [
    sortiment[0], # Milch
    sortiment[0], # Milch
    sortiment[2]  # Eier
]

gesamtpreis = 0

print("--- Warenkorb Details ---")
# Iteration über die Liste, bei der in jedem Durchlauf auf die Keys des aktuellen Dictionaries zugegriffen wird
for position in warenkorb:
    print("Artikel:", position["artikel"], "| Preis:", position["preis"], "EUR")
    gesamtpreis += position["preis"] # Aggregation der Einzelpreise

print("-----------------------")
print("Gesamtsumme:", gesamtpreis, "EUR")
```
