## 07 - Listen

Eine Liste ist eine veränderbare (*mutable*) Datenstruktur in Python, die dazu dient, mehrere Elemente in einer festgelegten Reihenfolge zu speichern. Listen werden durch eckige Klammern `[]` definiert.

Grundsätzlich können in einer Liste Elemente unterschiedlicher Datentypen (z. B. Strings, Integers, Booleans) gemischt werden. In der Praxis der professionellen Softwareentwicklung wird jedoch dringend empfohlen, pro Liste nur einen einheitlichen Datentyp zu verwenden.

```python
# Erstellen von Listen
stadtteile_norden = ["Pankow", "Reinickendorf", "Wedding"]
stadtteile_sueden = ["Schöneberg"]
noten = [4, 3, 2, 1]

# Gemischte Liste (technisch möglich, aber nicht empfohlen)
gemischte_liste = ["Bernburger Straße", True, 10963]

print("Liste:", stadtteile_norden)
```

### Zugriff, Indizierung und Slicing
Der Zugriff auf einzelne Listenelemente erfolgt über einen Index, der bei `0` beginnt.
* Ein negativer Index zählt vom Ende der Liste rückwärts (z. B. greift `-1` auf das letzte Element zu).
* Durch sogenanntes "List Slicing" (`liste[start:ende]`) lassen sich ganze Teilbereiche einer Liste extrahieren. Das End-Element ist dabei exklusiv.

```python
orte = ["Berlin", "Potsdam", "München", "Hamburg", "Köln"]

erstes_element = orte[0]
letztes_element = orte[-1]
drittletztes_element = orte[-3]

# Slicing: Extrahiert Elemente von Index 1 bis zum vorletzten Element. Letztes Element ist exklusive!
teil_liste = orte[1:-1]

print("Erstes Element:", erstes_element)
print("Letztes Element:", letztes_element)
print("Slicing-Ergebnis:", teil_liste)
```

### Elemente hinzufügen und entfernen
Listen bieten verschiedene Methoden, um Elemente dynamisch zur Laufzeit anzupassen:
* **`.append(element)`:** Fügt ein einzelnes Element am Ende der Liste hinzu.
* **`.extend(liste)`:** Erweitert die Liste um die Elemente einer anderen Liste. Alternativ kann auch der `+` Operator zur Listenverkettung genutzt werden.
* **`.insert(index, element)`:** Fügt ein Element an einer exakt definierten Position (Index) ein.
* **`.remove(wert)`:** Löscht das erste Vorkommen eines bestimmten Wertes aus der Liste.
* **`.pop(index)`:** Entfernt das Element am angegebenen Index und gibt diesen Wert als Rückgabewert zurück. Wird kein Index übergeben, wird standardmäßig das letzte Element (`-1`) entfernt.

```python
stadtteile = ["Kreuzberg", "Mitte"]

# Hinzufügen
stadtteile.append("Neukölln")
stadtteile.extend(["Schöneberg", "Zehlendorf"])
stadtteile.insert(0, "Charlottenburg")

print("Nach dem Hinzufügen:", stadtteile)

# Entfernen
stadtteile.remove("Schöneberg")
entferntes_element = stadtteile.pop() # Entfernt das letzte Element ("Zehlendorf")
spezifisch_entfernt = stadtteile.pop(2) # Entfernt das Element an Index 2

print("Nach dem Entfernen:", stadtteile)
print("Mit pop() entfernt:", entferntes_element)
```

### Listen durchsuchen, sortieren und kopieren
Weitere nützliche Methoden für die Arbeit mit Listen:
* **`.index(wert)`:** Gibt den Index des ersten Vorkommens eines bestimmten Wertes zurück.
* **`.count(wert)`:** Zählt, wie oft ein bestimmter Wert in der Liste vorkommt.
* **`.sort()`:** Sortiert die Elemente der Liste in aufsteigender (bzw. alphabetischer) Reihenfolge direkt im Speicher.
* **`.reverse()`:** Kehrt die aktuelle Reihenfolge der Listenelemente um.

**Kopieren von Listen (`.copy()`):**

Da Listen *mutable* (veränderbar) sind, führt eine einfache Zuweisung wie `liste_b = liste_a` dazu, dass beide Variablen auf **denselben Speicherort** verweisen. Änderungen an `liste_b` verändern somit auch `liste_a`. Um eine echte, unabhängige flache Kopie (Shallow Copy) zu erstellen, muss die Methode `.copy()` verwendet werden.

```python
zahlen = [4, 1, 2, 4, 3]

index_von_zwei = zahlen.index(2)
print("Die '2' besitzt den Index:", index_von_zwei)

anzahl_vieren = zahlen.count(4)
print("Die '4' kommt", anzahl_vieren, "mal in der Liste vor.")

zahlen.sort()
print("Sortierte Liste:", zahlen)

zahlen.reverse()
print("Umgekehrte Liste:", zahlen)

# Listen kopieren
original_liste = ["A", "B", "C"]
referenz_liste = original_liste             # Verweist auf den gleichen Speicherort!
unabhaengige_kopie = original_liste.copy()  # Erstellt eine unabhängige Kopie

original_liste.append("D")

print("Referenz (wurde automatisch mitgeändert):", referenz_liste)
print("Kopie (blieb unverändert):", unabhaengige_kopie)
```
