# Modul 3: Datenstrukturen
## 10 - Dictionaries

Ein Dictionary (Wörterbuch) ist eine veränderbare (*mutable*) Datenstruktur, die Daten nicht über einen fortlaufenden Index speichert, sondern über sogenannte Schlüssel-Wert-Paare (Key-Value-Paare). 

Dictionaries werden mit geschweiften Klammern `{}` definiert. Ein Dictionary ist prinzipiell unsortiert. Jeder Schlüssel (Key) muss innerhalb des Dictionaries eindeutig sein; wird einem existierenden Schlüssel ein neuer Wert zugewiesen, wird der alte Wert überschrieben.

```python
# Erstellen von Dictionaries
stadtteile = {"name": "Kreuzberg", "plz": 10969}

airports = {
    "Berlin": "BER", 
    "Helsinki": "HEL", 
    "Saigon": "SGN"
}

print("Dictionary 'stadtteile':", stadtteile)
```

### Zugriff, Hinzufügen und Entfernen
Der Lese- und Schreibzugriff auf die Elemente erfolgt über den entsprechenden Schlüssel in eckigen Klammern `[]`.

* **Sicherer Zugriff mit `.get()`:** Wird über `[]` auf einen Schlüssel zugegriffen, der nicht existiert, wirft Python einen `KeyError`. Die Methode `.get("Schlüssel")` verhindert dies und liefert stattdessen den Wert `None` (oder einen definierten Standardwert) zurück, falls der Schlüssel fehlt.
* **Elemente hinzufügen/ändern:** `dictionary["neuer_schluessel"] = wert`.
* **Elemente entfernen:** `del dictionary["schluessel"]` löscht ein spezifisches Paar. `.clear()` leert das gesamte Dictionary.

```python
stadtteile = {"name": "Kreuzberg", "plz": 10969}

# Hinzufügen und Ändern
stadtteile["einwohnerzahl"] = 150000
print("Nach Hinzufügen:", stadtteile)

stadtteile["einwohnerzahl"] = 250000 # Überschreibt den alten Wert
print("Nach Änderung:", stadtteile)

# Entfernen
del stadtteile["plz"]
print("Nach Löschen der PLZ:", stadtteile)

# Sicherer Lesezugriff
print("Direkter Zugriff:", airports["Berlin"])
print("Sicherer Zugriff (existiert):", airports.get("Berlin"))
print("Sicherer Zugriff (fehlt):", airports.get("Budapest"))    # Liefert None anstatt eines Fehlers
# print("Unsicherer Zugriff (fehlt):", airports["Budapest"])    # Liefert einen KeyError
```

### Iterieren über Dictionaries
Beim Durchlaufen eines Dictionaries mit einer Standard-`for`-Schleife (`for element in dict:`) werden standardmäßig nur die Schlüssel durchlaufen. 

Um gezielt auf Bestandteile zuzugreifen, bietet Python drei wichtige Methoden:
* **`.keys()`:** Liefert eine iterierbare Ansicht aller Schlüssel.
* **`.values()`:** Liefert eine iterierbare Ansicht aller Werte.
* **`.items()`:** Liefert Schlüssel und Werte gleichzeitig als Tupel zurück `(Key, Value)`. Dies wird sehr häufig in Kombination mit *Tuple Unpacking* im Schleifenkopf genutzt.

```python
# Mitgliedschaftsprüfung (in) bezieht sich standardmäßig auf Keys
print("Ist 'Berlin' ein Schlüssel?", "Berlin" in airports)
print("Ist 'BER' ein Wert?", "BER" in airports.values())

print("--- Iteration über Keys ---")
for schluessel in airports.keys():      # .keys() ist optional, for schluessel in airports: bewirkt dasselbe
    print("Schlüssel:", schluessel)

print("--- Iteration über Values ---")
for wert in airports.values():
    print("Wert:", wert)

print("--- Iteration über Items (Best Practice) ---")
for schluessel, wert in airports.items():
    print(schluessel + ": " + wert)

print("Umwandlung mit .items():", airports.items()) # Keine Liste im Arbeitspeicher! Lediglich ein View-Objekt auf das Dictionary.
```
