## 09 - Sets

Ein Set (Menge) ist eine ungeordnete Sammlung von eindeutigen Elementen. Sets werden primär genutzt, um Duplikate aus anderen Sequenzen (wie Listen) zu filtern oder um mathematische Mengenoperationen durchzuführen.

Sets werden mit geschweiften Klammern `{}` definiert, wobei doppelte Werte automatisch ignoriert werden.

```python
# Definition eines Sets (doppelte Werte werden ignoriert)
stadtteile_berlin_set = {"Kreuzberg", "Mitte", "Neukölln", "Mitte"}

# Ausgabe zeigt, dass "Mitte" nur einmal enthalten ist
print("Set:", stadtteile_berlin_set)

# Typische Anwendung: Duplikate aus einer Liste filtern
ort_liste = ["München", "Berlin", "München", "Hamburg"]
eindeutige_orte = list(set(ort_liste))

print("Gefilterte Liste:", eindeutige_orte)
```

### Elemente hinzufügen und entfernen
Da Sets ungeordnet sind, besitzen sie keinen festen Index. Ein Zugriff über `set[0]` ist demnach nicht möglich. Stattdessen werden Elemente direkt über ihren Wert angesprochen.
* **`.add(element)`:** Fügt ein neues Element hinzu.
* **`.remove(element)`:** Entfernt ein spezifisches Element (wirft einen Fehler, falls das Element nicht existiert).
* **`.discard(element)`:** Entfernt ein Element sicher (es wird kein Fehler geworfen, falls das Element nicht existiert).

```python
zahlen_set = {1, 2, 3}

zahlen_set.add(4)
zahlen_set.add(4)       # Wird ignoriert, da 4 bereits existiert

zahlen_set.remove(2)
zahlen_set.discard(10)  # Wirft keinen Fehler, obwohl 10 nicht existiert

print("Aktuelles Set:", zahlen_set)
```

### Mengenoperationen und Performance ($O(1)$)
Sets bieten effiziente Methoden für mathematische Vergleiche zwischen zwei Mengen:
* **Schnittmenge (`&`):** Findet Elemente, die in *beiden* Sets existieren.
* **Vereinigungsmenge (`|`):** Führt alle Elemente aus *beiden* Sets zusammen (Duplikate werden entfernt).
* **Differenz (`-`):** Findet Elemente, die nur im ersten Set, aber nicht im zweiten existieren.

**Performance-Vorteil:**
Die Überprüfung auf Mitgliedschaft (z. B. `"Wert" in set`) ist bei Sets deutlich performanter als bei Listen. Da Sets im Hintergrund als sogenannte Hash-Tabellen arbeiten, erfolgt die Prüfung in konstanter Zeit ($O(1)$). Bei einer Liste ($O(n)$) müsste im schlimmsten Fall jedes Element einzeln durchsucht werden.

```python
set_a = {"Max", "Erik", "Monika"}
set_b = {"Tobias", "Erik", "Anna"}

# Schnittmenge
print("Schnittmenge (in beiden):", set_a & set_b)

# Vereinigungsmenge
print("Vereinigung (alle):", set_a | set_b)

# Differenzmenge
print("Differenz (nur in A):", set_a - set_b)

# Differenzmenge
print("Differenz (nur in B):", set_b - set_a)

# Effiziente O(1) Mitgliedschaftsprüfung
print("Ist Erik in Set A?", "Erik" in set_a)

# Durch Typecasting von list zu set lassen sich die Operatoren auch auf Listen anwenden
liste_a = ["Max", "Erik", "Monika"]
liste_b = ["Tobias", "Erik", "Anna"]
schnittmenge = list(set(liste_a) & set(liste_b))
print("Schnittmenge aus 'Liste A' und 'Liste B':", schnittmenge)
```
