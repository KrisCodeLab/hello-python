## 11 - Queues und Itertools

Eine Queue (Warteschlange) ist eine Datenstruktur, die nach dem FIFO-Prinzip (First-In-First-Out) arbeitet. Elemente werden am Ende eingefügt und am Anfang wieder entnommen. Das integrierte Modul `queue` stellt diese Funktionalität in Python bereit.

```python
import queue

# Erstellen einer Standard-Queue
q = queue.Queue()

# Elemente hinzufügen
q.put("Hallo")
q.put("Welt")
q.put("Mars")
q.put("Pluto")

# Elemente entnehmen (striktes FIFO-Prinzip)
print("Erstes entnommenes Element:", q.get())
print("Zweites entnommenes Element:", q.get())

print("---")

# Restliche Elemente leeren und prüfen, ob die Queue leer ist
while not q.empty():
    print("Verbleibend:", q.get())
```

### PriorityQueue (Prioritätswarteschlange)
Eine `PriorityQueue` sortiert die Elemente nicht nach der Reihenfolge des Einfügens, sondern nach einer definierten Priorität. Elemente werden meist als Tupel in der Form `(Priorität, Wert)` übergeben. 

Das Element mit dem niedrigsten Zahlenwert besitzt hierbei die höchste Priorität und wird bei einem Aufruf von `.get()` als Erstes entnommen.

```python
pq = queue.PriorityQueue()

# Elemente mit Priorität hinzufügen (Priorität, Wert)
pq.put((10, "Hallo Welt"))
pq.put((15, "Mars"))
pq.put((5, "Wichtig")) # niedrigster Zahlenwert = höhste Priorität

# Element mit höchster Priorität (niedrigste Zahl) wird zuerst geholt
print("Höchste Priorität:", pq.get())
print("Zweithöchste Priorität:", pq.get())
```

### Kombinationen mit `itertools`
Das Modul `itertools` bietet Werkzeuge für effiziente Iterationen. Die Funktion `combinations(iterable, r)` generiert alle möglichen einzigartigen Kombinationen der Länge `r` aus den Elementen einer übergebenen Datenstruktur. 

Dies wird in der Praxis häufig für Optimierungsprobleme verwendet (z. B. wenn aus einer Einkaufsliste die optimale Gewichtsverteilung iterativ ermittelt werden soll, bei der systematisch verschiedene Kombinationen geprüft werden).

```python
import itertools

elemente = ["A", "B", "C", "D"]

# Alle möglichen 2er-Kombinationen aus der Liste generieren
kombinationen = list(itertools.combinations(elemente, 2))

print("Alle 2er-Kombinationen:")
for kombination in kombinationen:
    print(kombination)
```
