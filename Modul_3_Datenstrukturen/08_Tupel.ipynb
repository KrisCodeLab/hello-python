## 08 - Tupel

Ein Tupel ist eine weitere elementare Datenstruktur in Python. Es verhält sich grundsätzlich wie eine Liste, besitzt jedoch einen entscheidenden Unterschied: Tupel sind *immutable* (unveränderbar).
    
Tupel werden durch runde Klammern `()` definiert. Nach der Initialisierung können einem Tupel weder neue Elemente hinzugefügt, noch bestehende Elemente verändert oder entfernt werden. Dies schützt Datenstrukturen vor versehentlichen Modifikationen und verhält sich ähnlich wie feste Arrays in Sprachen wie Java.

```python
# Definition von Tupeln
stadtteile_berlin_tupel = ("Kreuzberg", "Mitte")
student = ("Max Müller", 22, "Informatik")
student_2 = ("Peter Meier", 23, "Germanistik")

print(stadtteile_berlin_tupel)
print("Datentyp:", type(student))

# Der Versuch einer Zuweisung (z.B. student[1] = 25) würde einen TypeError auslösen!
```

### Tuple Unpacking (Entpacken)
Ein sehr mächtiges und häufig genutztes Konzept in Python ist das sogenannte "Unpacking". Dabei lassen sich die einzelnen Elemente eines Tupels in einer einzigen Codezeile direkt separaten Variablen zuweisen[cite: 3]. 
    
Die Anzahl der Variablen auf der linken Seite muss dabei exakt der Anzahl der Elemente im Tupel entsprechen. Das Unpacking funktioniert auch bei Rückgabewerten von Funktionen, die mehrere Werte zurückgeben[cite: 3].

**Unpacking bei anderen Datenstrukturen:**
Das Prinzip des Entpackens ist in Python nicht auf Tupel beschränkt, sondern funktioniert bei allen iterierbaren (durchlaufbaren) Objekten wie Listen oder Strings. Solange die Anzahl der Variablen mit der Anzahl der Elemente übereinstimmt, werden die Werte indexbasiert in fester Reihenfolge zugewiesen. (Spezifische Verhaltensweisen beim Entpacken von Dictionaries werden im separaten Dictionary-Notebook behandelt).

```python
student = ("Max Müller", 22, "Informatik")

# Unpacking des Tupels in drei separate Variablen
name, age, subject = student

print("Name:", name)
print("Alter:", age)

# Unpacking einer Liste
plz, stadtteil = [10179, "Charlottenburg"]
print("PLZ aus Liste:", plz)

# Unpacking eines Strings
a, b, c = "Zug"
print("Buchstaben:", a, b, c)

# Unpacking bei Rückgabewerten von Funktionen
def get_student():
    return ["Peter Meier", 23, "Germanistik"]

name2, age2, subject2 = get_student()
print("Name aus Funktion:", name2)
```

### Tupel in Iterationen
Tupel werden in der Praxis häufig verwendet, um fest zusammengehörige Datensätze innerhalb einer Liste zu strukturieren. Beim Durchlaufen einer solchen Datenstruktur mit einer `for`-Schleife lässt sich das Unpacking direkt im Schleifenkopf anwenden, was den Code sehr kompakt und lesbar macht.

```python
# Liste, die aus mehreren Tupeln besteht
students = [
    ("Max Müller", 22),
    ("Peter Meier", 23)
]

# Iteration mit direktem Unpacking im Schleifenkopf
for name, age in students:
    print("Student:", name, "| Alter:", age)
```
