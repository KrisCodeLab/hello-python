## 02 - Variablen und Datentypen

Python ist eine dynamisch typisierte Programmiersprache. Das bedeutet, dass der Datentyp einer Variable bei der Zuweisung automatisch erkannt wird und nicht bei der Instanziierung festgelegt werden muss. Somit können Variablen auch einen anderen Datentyp zugewiesen bekommen, als ursprünglich erhalten.

Die wichtigsten primitiven Datentypen in Python sind:
* **Integer (int):** Ganze Zahlen (z. B. `6`, `-56`).
* **Float (float):** Fließkommazahlen. Hier wird immer ein Punkt statt eines Kommas verwendet (z. B. `6.0`).
* **String (str):** Zeichenketten. Sie können mit einfachen oder doppelten Anführungszeichen definiert werden (z. B. `'Bernburger Straße'` oder `"Matthias 'Matze' Gottfried"`)
* **Boolean (bool):** Wahrheitswerte, die exakt zwei Zustände annehmen können: `True` oder `False`.

```python
# Variablen und ihre Zuweisung
ganze_zahl_integer = 6
negative_fliesskomma_zahl = -56.5

strasse = "Bernburger Straße"
name = "Matthias 'Matze' Gottfried"

ergebnis = True
bestanden = False
```

### Typen prüfen (`type` und `isinstance`)
Da sich der Typ einer Variable zur Laufzeit ändern kann, bietet Python integrierte Funktionen, um Datentypen zu überprüfen. 
* `type()` liefert den exakten Datentyp eines Wertes oder einer Variable zurück.
* `isinstance()` prüft, ob eine Variable einem bestimmten Datentyp entspricht und gibt einen Boolean (`True` oder `False`) zurück. Dies ist die bevorzugte Methode im professionellen Clean Code.

```python
# Datentyp ermitteln
print(type(strasse))
print(type(ganze_zahl_integer))

# Datentyp gezielt testen (Liefert True oder False)
print(isinstance(ganze_zahl_integer, int))
print(isinstance(negative_fliesskomma_zahl, float))
```

### Type Casting (Typumwandlung)
Oft müssen Daten von einem Typ in einen anderen umgewandelt werden, zum Beispiel wenn Benutzereingaben (die immer als String ankommen) für Berechnungen genutzt werden sollen. Python nutzt dafür einfache Funktionen:
* `int()` wandelt einen kompatiblen Wert in eine Ganzzahl um.
* `float()` erzeugt eine Fließkommazahl.
* `str()` macht aus einem Wert eine Zeichenkette (String).

```python
# String nach Integer
zahl_1 = int("1000")
zahl_2 = int("1000") + int("500")

# Integer nach Float
gebrochene_zahl = float(1000)

# Integer nach String
string = str(zahl_2)

print("Berechnete Zahl:", zahl_2)
print("Als Float:", gebrochene_zahl)
print("Als Stringverkettung:", string + " Äpfel")
```

### Type Hinting (PEP 484)
Obwohl Python dynamisch typisiert ist, schreibt man in modernen, professionellen Anwendungen oft sogenannte "Type Hints" (Typ-Hinweise). Sie erzwingen zwar keinen Datentyp, helfen aber der Entwicklungsumgebung und anderen Entwicklern, Fehler schneller zu finden.

```python
# Variable: Typ = Wert
alter: int = 25
gehalt: float = 3500.50
ist_eingeloggt: bool = True
benutzername: str = "Admin"
```

### Mutability (Veränderbarkeit)
Ein extrem wichtiges Konzept für die Speichereffizienz: In Python ist alles ein Objekt im Arbeitsspeicher. Datentypen in Python sind entweder *mutable* (veränderbar) oder *immutable* (unveränderbar).
* **Immutable (Unveränderbar):** Integers, Floats, Booleans und Strings. Beispiel: Bei veränderung eines Strings, erstellt Python im Hintergrund immer ein komplett neues Objekt im Arbeitsspeicher.
* **Mutable (Veränderbar):** Listen oder Dictionaries. Sie können direkt im Speicher modifiziert werden.

Wird der Wert einer veränderbaren Variable geändert, wird diese nicht überschrieben, sondern an einem anderen Speicherplatz als neues Objekt erstellt. Den Speicherort kann man sich mit `hex(id(variable))` anzeigen lassen.

```python
a = 5
print("Speicherort von a (Wert 5):", hex(id(a)))

# Der Wert der variable wird verändert:
a = 10
# Tatsächlich hat Python ein neues Objekt im Speicher angelegt:
print("Neuer Speicherort von a (Wert 10):", hex(id(a)))
```
