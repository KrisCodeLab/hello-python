### Namenskonventionen (PEP 8 Style Guide)
Bei der Benennung von Bezeichnern gelten in Python feste Konventionen:
- **snake_case:** Standard für Variablen und Funktionen (z. B. `ganze_zahl`). Wörter werden kleingeschrieben und mit Unterstrichen getrennt.
- **PascalCase:** Standard für Klassennamen (z. B. `FahrzeugModell`).
- **camelCase:** In Python unüblich (wird primär in Java oder JavaScript genutzt).
- **Regeln:** Namen dürfen niemals mit einer Ziffer beginnen oder Leerzeichen enthalten. Umlaute sind technisch möglich, führen aber zu Kompatibilitätsproblemen und gehören nicht in sauberen Code.
- Ein führender Unterstrich (z. B. `_interne_zahl`) signalisiert nach Konvention interne bzw. private Variablen.

```python
# Zulässige und empfohlene Bezeichner (PEP 8 Standard):
ganze_zahl = 45
GanzeZahl = 123
_ganze_zahl = 5

# Technisch möglich, aber unbedingt vermeiden (kein Clean Code):
häuserzeile = "Hallo"

# Ungültige Bezeichner (SyntaxError):
# 1ganzeZahl = 7
# ganze zahl = 8
```

### Zuweisungen & Lesbarkeit (Clean Code)
Code wird wesentlich öfter gelesen als geschrieben. Mehrfachzuweisungen in einer einzelnen Zeile sind syntaktisch zwar erlaubt, erschweren jedoch das Debugging und die Lesbarkeit und gelten daher als schlechter Stil.

```python
# Best Practice: Eindeutige, separate Zuweisung
zahl = 1500

# Schlechter Stil (Unnötig verschachtelt):
ergebnis = zahl = 1500
```

### Kommentare vs. Docstrings
- **Einzeilige Kommentare:** Beginnen mit `#`.
- **Docstrings:** Werden in dreifache Anführungszeichen `"""` gefasst. Sie dienen der formalen Dokumentation direkt unterhalb von Modulen, Klassen oder Funktionen und können über `help()` abgerufen werden.

```python
# Das ist ein regulärer einzeiliger Kommentar

"""
Das ist ein mehrzeiliger String / Docstring.
Er wird primär genutzt, um Funktionen oder Klassen strukturiert zu dokumentieren.
"""

def beispiel_funktion():
    """Diese Funktion dient als Dokumentationsbeispiel."""
    pass

help(beispiel_funktion)
```
