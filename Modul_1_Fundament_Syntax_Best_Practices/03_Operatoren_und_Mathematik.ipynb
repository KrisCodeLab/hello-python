## 03 - Operatoren und Mathematik

In Python gibt es verschiedene Kategorien von Operatoren, um mathematische Berechnungen durchzuführen, Werte zu vergleichen oder Bedingungen logisch zu verknüpfen.

### Arithmetische Operatoren
Neben den klassischen Grundrechenarten (+, -, *, /) stehen in Python erweiterte mathematische Operatoren zur Verfügung:
* **Modulo (`%`):** Gibt den Restwert einer Division zurück.
* **Potenz (`**`):** Berechnet die Potenz (z. B. "hoch 3").
* **Ganzzahlige Division (`//`):** Teilt zwei Zahlen und schneidet die Nachkommastellen des Ergebnisses ab (liefert einen Integer).

```python
variable_1 = 10.0
variable_2 = 3

summe = variable_1 + variable_2
differenz = variable_1 - variable_2
produkt = variable_1 * variable_2
division = variable_1 / variable_2

# Erweiterte arithmetische Operatoren
restwert = variable_1 % variable_2
potenzwert = variable_1 ** variable_2
ganzzahlige_division = variable_1 // variable_2

print("Restwert:", restwert)
print("Potenzwert:", potenzwert)
print("Ganzzahlige Division:", ganzzahlige_division)

# Bekanntes Phänomen bei Fließkommazahlen (Ungenauigkeit im Binärsystem)
ungenau = 0.3 + 0.3 + 0.3
print("Ungenaues Binär-Ergebnis:", ungenau)

# Behebung der Ungenauigkeit:
korrigiert = round(ungenau, 2)
print("Korrigiert durch Runden:", korrigiert)
```

### Zuweisungsoperatoren
Um Code kürzer und lesbarer zu gestalten, lässt sich die Zuweisung von Variablen mit arithmetischen Operatoren kombinieren. Statt der langen Schreibweise wird oft die kurze Schreibweise (z. B. `+=`) verwendet.

```python
preis = 15

# Lange Schreibweise
preis = preis + 3

# Kurze Schreibweise (Addiert 3 zum aktuellen Wert)
preis += 3

# Funktioniert auch mit anderen Operatoren (z.B. -=, *=, /=)
preis *= 2
```

### Vergleichs- und Identitätsoperatoren
Vergleichsoperatoren prüfen das Verhältnis zweier Werte zueinander. Das Ergebnis einer solchen Prüfung ist immer ein Boolean (`True` oder `False`). 
* **Gleichheit / Ungleichheit:** `==` prüft auf exakte Gleichheit, `!=` auf Ungleichheit.
* **Größer / Kleiner:** `<`, `>`, `<=`, `>=`.
* **Identität:** `is not` oder `is` vergleicht, ob zwei Variablen auf dasselbe Objekt im Speicher verweisen (nicht nur auf inhaltliche Gleichheit).

```python
ort1 = "Berlin"
ort2 = "Paris"

print("Gleichheit:", ort1 == ort2)
print("Kleiner als:", ort1 < ort2) # Alphabetischer Vergleich: B kommt vor P

ergebnis1 = 0.3 * 3
ergebnis2 = 0.9
ergebnis3 = ergebnis2

print("Ungleichheit:", ergebnis1 != ergebnis2)
print("Identitätsprüfung:", ergebnis1 is not ergebnis2)
print("Identitätsprüfung:", ergebnis3 is ergebnis2)     # True da ergebnis3 auf ergebnis2 zeigt
```

### Mitgliedschaftsoperatoren (`in` / `not in`)
Mitgliedschaftsoperatoren prüfen, ob ein spezifisches Element Teil einer Sequenz ist. Dies funktioniert sowohl bei Zeichenketten (Strings) als auch bei Datenstrukturen wie Listen oder Dictionaries.

```python
# Mitgliedschaft in einem String prüfen
ort = "Strasbourg"
print("a in Strasbourg:", "a" in ort)

# Mitgliedschaft in einer Liste prüfen
staedte = ["Berlin", "Karlsruhe", "Freiburg"]
print("Berlin in staedte:", "Berlin" in staedte)
print("Ber in staedte:", "Ber" in staedte)          # False, gesucht wird das exakte Element
print("Hamburg nicht in staedte:", "Hamburg" not in staedte)
```

### Logische Operatoren (`and`, `or`, `not`)
Um komplexe Abfragen zu erstellen, lassen sich mehrere Bedingungen miteinander verknüpfen:
* **`and`:** Liefert `True`, wenn *alle* Bedingungen wahr sind.
* **`or`:** Liefert `True`, wenn *mindestens eine* Bedingung wahr ist.
* **`not`:** Kehrt den Wahrheitswert um (aus `True` wird `False` und umgekehrt).

```python
# Logische Verknüpfungen
temperatur = 25
wetter = "sonnig"

# and-Verknüpfung
gutes_wetter = temperatur > 20 and wetter == "sonnig"

# or-Verknüpfung
regenschirm_noetig = wetter == "regnerisch" or wetter == "bewölkt"

# not-Umkehrung
print("Ist das Wetter nicht gut?", not gutes_wetter)

# Logisches XOR (Entweder-Oder) mithilfe von !=
bedingung_1 = True
bedingung_2 = False

xor_ergebnis = bedingung_1 != bedingung_2
print("XOR-Logik erfüllt (entweder 1 oder 2 ist wahr)?", xor_ergebnis)
```
