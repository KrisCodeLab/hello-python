## 04 - Strings und Formatierung

Zeichenketten (Strings) dienen der Speicherung und Verarbeitung von Text. Ein String wird durch einfache oder doppelte Anführungszeichen definiert.

### Verkettung und Multiplikation
Strings lassen sich mit mathematischen Operatoren manipulieren:
* Der Operator `+` verbindet (verkettet) zwei Strings miteinander.
* Der Operator `*` in Kombination mit einem Integer wiederholt einen String entsprechend oft.

```python
string_1 = "Hallo aus "
string_2 = "München"
multiplikator = 3

verkettung = string_1 + string_2
verkettung_2 = string_2 * multiplikator

print("Verkettung:", verkettung)
print("Wiederholung:", verkettung_2)
```

### Indizierung und Slicing
Strings sind sequenzielle Datentypen. Jedes Zeichen besitzt einen festen Index, beginnend bei `0`.
* Der Abruf eines einzelnen Zeichens erfolgt über eckige Klammern, z. B. `string[2]`.
* Ein negativer Index zählt vom Ende des Strings rückwärts (z. B. greift `-1` auf das letzte Zeichen zu).
* Durch sogenanntes "Slicing" lassen sich ganze Teilbereiche extrahieren. Die Syntax lautet `string[start:ende]`, wobei das angegebene Ende exklusiv ist.

```python
ort = "Strasbourg"

zweiter_buchstabe = ort[2]
letzter_buchstabe = ort[-1]

# String-Slicing:
ersten_vier = ort[0:4]
teilstring = ort[1:-1]

print("Zweiter Buchstabe (Index 2):", zweiter_buchstabe)
print("Letzter Buchstabe:", letzter_buchstabe)
print("Ersten vier Buchstaben:", ersten_vier)
print("Slicing von Index 1 bis zum vorletzten Zeichen:", teilstring)
```

### Wichtige String-Methoden
Da in Python alles ein Objekt ist, bringen Strings fest eingebaute Funktionen (Methoden) mit, um den Text zu manipulieren. Da Strings *immutable* (unveränderbar) sind, geben diese Methoden immer einen **neuen** String zurück, anstatt den alten im Speicher zu überschreiben.

* **`.strip()`**: Entfernt Leer- und Zeilenumbruchszeichen am Anfang und Ende des Strings.
* **`.replace(alt, neu)`**: Ersetzt definierte Bestandteile des Strings.
* **`.split(trennzeichen)`**: Zerteilt den String anhand eines Trennzeichens und liefert eine Liste zurück.
* **`"trennzeichen".join(liste)`**: Verbindet Elemente einer Liste zu einem einzigen String.
* **`.upper()` / `.lower()`**: Wandelt den Text in Groß- bzw. Kleinbuchstaben um.
* **`.find(zeichen)`**: Sucht nach einem Zeichen und liefert den Index zurück. Gibt `-1` aus, falls das Zeichen nicht gefunden wird.

```python
# Bereinigung und Ersetzung
text = "   Hallo Welt.   "
bereinigt = text.strip()
ersetzter_text = bereinigt.replace("Welt", "Python")

# Splitten und Zusammenfügen
orte_als_string = "München, Berlin, Potsdam"
orte_liste = orte_als_string.split(", ")
orte_pfeil = " -> ".join(orte_liste)

print("Bereinigt:", bereinigt)
print("Upper Case:", bereinigt.upper())
print("Ersetzt:", ersetzter_text)
print("Als Liste:", orte_liste)
print("Verbunden:", orte_pfeil)
print("1. '->' suchen:", orte_pfeil.find("-"))
```

### String-Formatierung
Um Variableninhalte dynamisch in Texte einzufügen, bietet Python verschiedene Möglichkeiten.
* **f-Strings (Format-Strings):** Werden mit einem vorangestellten `f` gekennzeichnet. Variablen können direkt in geschweiften Klammern `{}` eingefügt werden. Dies ist der moderne und empfohlene Standard.
* **`.format()`-Methode:** Eine etwas ältere Methode, bei der geschweifte Klammern als Platzhalter fungieren.
* **`%`-Formatierung:** Eine veraltete Methode, die an die Programmiersprache C angelehnt ist (z. B. `%i` für Integer, `%s` für String).
* **Formatierungs-Optionen:** Innerhalb der Klammern können durch einen Doppelpunkt `:` zusätzliche Formatierungen angewendet werden, wie z. B. die Begrenzung von Nachkommastellen (`.2f`), feste Textbreiten (`10`), Füllzeichen (`*<30`) oder Zentrierung (`^`).

```python
postleitzahl = 10963
ort = "Kreuzberg"
parkanteil = 14.5692

# f-String (Moderner Standard)
print(f"Hallo {postleitzahl} in {ort}! Parkanteil: {parkanteil}")

# f-String mit Limitierung auf 2 Nachkommastellen (.2f)
print(f"Hallo {postleitzahl} in {ort}! Parkanteil: {parkanteil:.2f}")

# .format() Methode
print("Hallo {} in {}! Parkanteil: {}".format(postleitzahl, ort, parkanteil))

# Ausrichtung und Auffüllen mit f-Strings und .format()
print(f"Hallo {postleitzahl:10} in {ort:*<30}! Parkanteil: {parkanteil:.3f}")
print("Hallo {:^10} in {:<15}! Parkanteil: {:.2f}".format(postleitzahl, ort, parkanteil))

# Bedingte Formatierung innerhalb eines f-Strings
print(f"Hallo {postleitzahl}! {'Hoher' if parkanteil > 10 else 'Niedriger'} Parkanteil.")
```
