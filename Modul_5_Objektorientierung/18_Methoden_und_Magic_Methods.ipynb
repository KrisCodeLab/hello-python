## 18 - Methoden und Magic Methods

Klassen umfassen neben Eigenschaften (Variablen) auch Funktionen, die das Verhalten eines Objekts definieren. Diese an ein Objekt gebundenen Funktionen werden als **Methoden** bezeichnet. 

Wie beim Konstruktor (`__init__`) wird bei jeder regulären Methode das Schlüsselwort `self` als erster Parameter übergeben. Dies stellt sicher, dass die Methode auf die Daten der spezifischen Instanz zugreift.

```python
class Fahrzeug:
    def __init__(self, marke, geschwindigkeit):
        self.marke = marke
        self.geschwindigkeit = geschwindigkeit

    # Definition einer regulären Methode
    def beschleunigen(self, pace):
        print("Fahrzeug wird beschleunigt...")
        self.geschwindigkeit += pace
        print("Neue Geschwindigkeit:", self.geschwindigkeit, "km/h\n")

# Instanziierung und Methodenaufruf
mein_auto = Fahrzeug("BMW", 0)
mein_auto.beschleunigen(15)
```

### Magic Methods (Dunder-Methoden)
Python verfügt über spezielle Methoden, die mit doppelten Unterstrichen beginnen und enden (Double Underscore = "Dunder"). Diese Methoden werden meist nicht explizit aufgerufen, sondern automatisch im Hintergrund ausgeführt, wenn bestimmte Standard-Operationen stattfinden.

* **`__str__(self)`:** Wird ausgeführt, wenn ein Objekt mit `print()` ausgegeben oder per `str()` in Text umgewandelt wird. Ohne diese Methode wird lediglich der kryptische Speicherort des Objekts ausgegeben. Dies ist vergleichbar mit der `toString()`-Methode in Java.
* **`__eq__(self, other)`:** Kommt zum Einsatz, wenn zwei Objekte mit dem Operator `==` verglichen werden. Ohne diese Methode wird standardmäßig nur geprüft, ob es sich um exakt dieselbe Speicherreferenz handelt.

```python
class Fahrzeug:
    def __init__(self, marke, farbe):
        self.marke = marke
        self.farbe = farbe

    # Überschreiben der Standard-Textausgabe
    def __str__(self):
        fahrzeuginfo = self.marke + " / Farbe: " + self.farbe
        return fahrzeuginfo

    # Überschreiben des Vergleichs-Operators (==)
    def __eq__(self, fahrzeug_instanz):
        # Zwei Fahrzeuge gelten hier als "gleich", wenn ihre Farbe identisch ist
        if self.farbe == fahrzeug_instanz.farbe:
            return True
        return False

auto1 = Fahrzeug("VW", "schwarz")
auto2 = Fahrzeug("BMW", "schwarz")

# Test der __str__ Methode (wird durch print() automatisch getriggert)
print("Ausgabe auto1:", auto1)

# Test der __eq__ Methode (wird durch == automatisch getriggert)
print("Sind die Fahrzeuge farblich identisch?", auto1 == auto2)
```
