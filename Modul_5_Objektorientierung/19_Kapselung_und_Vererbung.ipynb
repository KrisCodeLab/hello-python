### Kapselung (Encapsulation) und Properties
Kapselung verhindert den direkten Lese- oder Schreibzugriff auf den internen Zustand eines Objekts von außen. In Python wird eine Variable durch zwei vorangestellte Unterstriche (z. B. `__term`) "privatisiert" (Name Mangling). 

Um dennoch kontrolliert auf diese privaten Variablen zugreifen zu können, werden Getter- und Setter-Methoden verwendet. Die sogenannten Dekoratoren `@property` (für den Lesezugriff) und `@variablenname.setter` (für den Schreibzugriff) ermöglichen eine elegante Syntax: Sie lassen Methoden nach außen wie normale Variablenaufrufe wirken, führen aber intern Validierungslogiken aus.

```python
class Student:
    def __init__(self, firstname, lastname):
        self.firstname = firstname
        self.lastname = lastname
        self.__term = 1 # Privates Attribut, von außen nicht direkt änderbar

    @property
    def term(self):
        """Getter: Wird aufgerufen, wenn der Wert gelesen werden soll."""
        return self.__term

    @term.setter
    def term(self, value):
        """Setter: Wird aufgerufen, wenn dem Attribut ein Wert zugewiesen wird."""
        if value > 9:
            print("Fehler: Ungültiger Wert für Semester. Wert wurde nicht geändert.")
            return
        self.__term = value

    def info(self):
        return f"{self.firstname} {self.lastname} (Semester: {self.__term})"

# Test der Kapselung
erik = Student("Erik", "Mustermann")

# Aufruf des Setters (sieht aus wie eine normale Variablenzuweisung)
erik.term = 10
erik.term = 5

# Aufruf des Getters
print("Aktuelles Semester:", erik.term)
print(erik.info())
```

### Vererbung (Inheritance) und `super()`
Durch Vererbung können neue Klassen (Unterklassen/Subclasses) erstellt werden, die alle Eigenschaften und Methoden einer bereits bestehenden Klasse (Oberklasse/Superclass) übernehmen. Dies verhindert redundanten Code (DRY-Prinzip).

Die Unterklasse kann:
1. Neue, spezifische Attribute und Methoden hinzufügen.
2. Bestehende Methoden der Oberklasse überschreiben (Overriding).

Die Funktion `super()` wird verwendet, um aus der Unterklasse heraus explizit auf die Methoden der Oberklasse zuzugreifen. Dies wird fast immer im Konstruktor (`__init__`) angewendet, um die geerbten Grundattribute ordnungsgemäß zu initialisieren.

```python
# Die Klasse WorkingStudent erbt alle Eigenschaften der Klasse Student
class WorkingStudent(Student):

    def __init__(self, firstname, lastname, company):
        # super() ruft den Konstruktor der Oberklasse auf
        super().__init__(firstname, lastname)
        # Spezifische Eigenschaft, die nur ein WorkingStudent besitzt
        self.company = company

    # Überschreiben (Overriding) der info-Methode der Oberklasse
    def info(self):
        # Holt den formatierten Namen der Oberklasse und hängt die Firmeninfo an
        basis_info = super().info()
        return f"{basis_info} - Arbeitet bei: {self.company}"

# Instanziierung der Unterklasse
peter = WorkingStudent("Peter", "Meyer", "Tech GmbH")

# peter besitzt Methoden und Attribute beider Klassen
print(peter.info())
print("Ist peter ein WorkingStudent?", isinstance(peter, WorkingStudent))
print("Ist peter auch ein Student?", isinstance(peter, Student))
```
