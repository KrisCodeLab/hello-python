## 17 - Klassen, Objekte und der Konstruktor

In der objektorientierten Programmierung (OOP) werden reale Sachverhalte modelliert. Anstatt Variablen und Funktionen lose im Code zu belassen, werden diese logisch in **Klassen** gebündelt. 

Eine Klasse (`class`) fungiert als abstrakter Bauplan. Das **Objekt** (auch Instanz genannt) ist das konkrete, nach diesem Bauplan erstellte Konstrukt im Arbeitsspeicher.

```python
# Definition einer einfachen Klasse
class Student:
    pass

# Erstellen von zwei konkreten Objekten (Instanzen) aus dem Bauplan
erik = Student()
monika = Student()

print("Objekt erik:", type(erik))
print("Objekt monika:", type(monika))
```

### Der Konstruktor (`__init__`) und `self`
Um einem Objekt direkt bei der Erstellung individuelle Eigenschaften (Attribute) mitzugeben, wird der sogenannte **Konstruktor** genutzt. In Python ist dies die spezielle Methode `__init__`.

* **`self`:** Dieses Schlüsselwort ist von zentraler Bedeutung. Es zeigt immer als Speicherreferenz auf das *eigene, konkrete Objekt*, das gerade erstellt oder bearbeitet wird. Es wird als erster Parameter in jeder Objekt-Methode übergeben, damit das Objekt intern auf seine eigenen Daten (z. B. `self.firstname`) zugreifen kann.

```python
class Student:
    # Der Konstruktor wird automatisch aufgerufen, wenn ein Objekt erstellt wird
    def __init__(self, firstname, lastname):
        self.firstname = firstname
        self.lastname = lastname
        self.term = 1

    def name(self):
        # self garantiert, dass hier der Name DIESES spezifischen Objekts gelesen wird
        return f"{self.firstname} {self.lastname} (Semester: {self.term})"

    def increase_term(self):
        if self.term >= 9:
            return
        self.term += 1

# Objekte mit individuellen Eigenschaften initialisieren
erik = Student("Erik", "Mustermann")
monika = Student("Monika", "Müller")

erik.increase_term()

print(erik.name())
print(monika.name())
```

### Klassenvariablen vs. Instanzvariablen
Es existiert ein fundamentaler Unterschied bezüglich des Definitionsortes von Variablen:

1. **Instanzvariablen:** Werden im Konstruktor (`__init__`) mit `self.` definiert. Jedes Objekt besitzt eigene, unabhängige Werte für diese Variablen.
2. **Klassenvariablen (Statische Variablen):** Werden direkt in der Klasse definiert, *außerhalb* des Konstruktors. Sie gelten global für *alle* Objekte dieser Klasse gleichermaßen. Nach gängiger Konvention gehören individuelle Eigenschaften von Objekten daher immer in den Konstruktor.

```python
class Car:
    # Klassenvariable: Gilt für alle Instanzen dieser Klasse
    price = "expensive"

# Objekt erstellen
bmw = Car()
dacia = Car()

print("Objekt-Preis (bezieht sich auf die Klasse):", bmw.price)
print("Klassen-Preis direkt abgerufen:", Car.price)

print("\n--- Nach Änderung der Klassenvariable ---")

# Die Variable wird global für die gesamte Klasse geändert
Car.price = "more expensive"

# Das Objekt bmw greift nun auf den neuen globalen Wert zu
print("Neuer BMW-Preis:", bmw.price)
print("Neuer Dacia-Preis:", dacia.price)

# Die Klassenvariable kann auch für ein einzelnes Objekt geändert werden
print("\n--- Nach Änderung der Klassenvariable im Objekt ---")

dacia.price = "cheap"

print("BMW-Preis unverändert:", bmw.price)
print("Neuer Dacia-Preis:", dacia.price)

# Durch das shadowing der Objektvariable, bleibt dacia durch weitere Änderung der Instanzvariable unberührt

print("\n--- Nach erneuter Änderung der Klassenvariable ---")

Car.price = "gratis"

print("Neuer BMW-Preis:", bmw.price)
print("Dacia-Preis unverändert:", dacia.price)

```
