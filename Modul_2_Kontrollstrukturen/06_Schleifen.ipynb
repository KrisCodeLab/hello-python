## 06 - Schleifen (while, for, range & Steuerungsanweisungen)

Schleifen (Loops) dienen dazu, bestimmte Code-Blöcke mehrfach auszuführen, ohne den Code manuell wiederholen zu müssen. In Python gibt es zwei primäre Arten von Schleifen:
* **`for`-Schleife:** Iteriert über eine Sequenz (z. B. eine Liste, ein Tupel oder eine Zeichenkette).
* **`while`-Schleife:** Führt den Code-Block so lange aus, wie eine definierte logische Bedingung `True` ergibt. Hier muss sichergestellt werden, dass sich die Bedingung im Laufe der Schleife ändert, um eine Endlosschleife zu vermeiden.

```python
stadtbezirke = ["Schwabing", "Solln", "Sendling", "Moosach"]

# for-Schleife
for name in stadtbezirke:
    print(name, "ist ein Stadtbezirk von München")

print("---")

# while-Schleife
element_index = 0
while element_index < len(stadtbezirke):
    name = stadtbezirke[element_index]
    print(name, "ist ein Stadtbezirk von München")
    element_index += 1
```

### Die Funktion `range()`
Die integrierte Funktion `range()` generiert eine Sequenz von Zahlen und wird verwendet, um `for`-Schleifen eine bestimmte Anzahl von Durchläufen vorzugeben. 
Die Syntax lautet `range(Start, Stopp, Schrittweite)`, wobei der Stopp-Wert stets exklusiv (nicht eingeschlossen) ist.

```python
# Zählen von 0 bis 4 (Stopp-Wert 5 ist exklusiv)
for nummer in range(0, 5):
    print(nummer)

print("---")

# Rückwärts zählen mit negativer Schrittweite
for nummer in range(5, 0, -2):
    print(nummer)
```

### Steuerungsanweisungen (`break`, `continue`, `pass`)
Der Ablauf einer Schleife lässt sich durch spezifische Anweisungen gezielt steuern:
* **`break`:** Bricht die gesamte Schleife sofort ab.
* **`continue`:** Bricht nur den aktuellen Schleifendurchlauf ab und springt sofort zum Beginn des nächsten Durchlaufs.
* **`pass`:** Dient als reiner syntaktischer Platzhalter. Es passiert nichts, aber es verhindert einen Syntax-Fehler, wenn ein Code-Block erwartet wird.

```python
# break
for name in stadtbezirke:
    if name == "Solln":
        break
    print("Bearbeite:", name) # Beendet die Schleife bei Solln

print("---")

# continue
for name in stadtbezirke:
    if name == "Sendling":
        continue
    print("Bearbeite Daten für", name) # Sendling wird übersprungen

# pass (Platzhalter)
for name in stadtbezirke:
    pass
```

### Der `else`-Block bei Schleifen
Eine Besonderheit in Python ist, dass Schleifen (`for` und `while`) mit einem `else`-Block kombiniert werden können. 
Der Code im `else`-Block wird **nur dann** ausgeführt, wenn die Schleife regulär und vollständig durchlaufen wurde. Wurde die Schleife durch ein `break` vorzeitig abgebrochen, wird der `else`-Block ignoriert.

```python
# Regulärer Durchlauf: else wird ausgeführt
for eintrag in stadtbezirke:
    print(eintrag)
    if name == "Hasenbergl":
        break
else:
    print("Liste regulär abgearbeitet.")

print("---")

# Abbruch durch break: else wird NICHT ausgeführt
for eintrag in stadtbezirke:
    print(eintrag)
    break
else:
    print("Wird nicht gedruckt, da die Schleife durch break beendet wurde.")
```

### Fortgeschrittenes Iterieren (`zip` und `enumerate`)
Um das Iterieren über Datenstrukturen effizienter zu gestalten, bietet Python hilfreiche Funktionen:
* **`zip()`:** Verbindet zwei oder mehr iterierbare Objekte (z. B. Listen) indexbasiert zu Tupeln. Ideal, um zwei Arrays parallel zu durchlaufen.
* **`enumerate()`:** Fügt einer Iteration automatisch einen Zähler (Index) hinzu. Dies erspart das manuelle Definieren und Hochzählen einer Zählervariable (wie `element_index += 1` bei einer `while`-Schleife).

```python
plz = [10967, 10365, 10719]
stadtteile = ["Kreuzberg", "Lichtenberg", "Wilmersdorf"]

# zip(): Zwei Listen parallel durchlaufen
stadtteile_mit_plz = list(zip(plz, stadtteile))

print(stadtteile_mit_plz)

for plz_eintrag, stadtteil_eintrag in stadtteile_mit_plz:
    print(f"PLZ: {plz_eintrag}, Ort: {stadtteil_eintrag}")

print("---")

# enumerate(): Index und Wert gleichzeitig abrufen
for index, stadtteil in enumerate(stadtteile):
    print(f"Index {index} gehört zu {stadtteil}")
```
