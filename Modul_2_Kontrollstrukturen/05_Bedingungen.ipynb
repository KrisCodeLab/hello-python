# Modul 2: Kontrollstrukturen
## 05 - Bedingungen (if, elif, else & match)

In Python werden Kontrollstrukturen genutzt, um den Programmablauf zu steuern. Die `if`-Anweisung führt einen Codeblock nur dann aus, wenn eine bestimmte Bedingung erfüllt (also `True`) ist.

Ergänzend kann mit `else` ein alternativer Codeblock definiert werden, der ausgeführt wird, falls die vorherige Bedingung nicht zutrifft. 
Wichtig: In Python wird die Zugehörigkeit von Codeblöcken strikt durch Einrückung (Indentation) definier.

```python
m = 10

if m < 5:
    print("m ist kleiner als 5")
else:
    print("Bedingung ist nicht erfüllt")
```

### Mehrfache Bedingungen (`elif`) und Verschachtelung
Um mehrere Bedingungen nacheinander zu prüfen, verwendet man `elif` (kurz für "else if"). Sobald eine Bedingung in der Kette erfüllt ist, wird der entsprechende Block ausgeführt und die Struktur verlassen.

Zudem lassen sich `if`-Bedingungen auch ineinander verschachteln. 

Ein nützliches Schlüsselwort in diesem Kontext ist `pass`: Damit kann eine Bedingung (z. B. ein `else`-Block) explizit leer gelassen werden, ohne dass ein Syntax-Fehler entsteht.

```python
plz = 10969
ort = "Potsdam"

if ort == "Berlin":
    print(ort)
    if plz == 10969:
        print("Beide Bedingungen erfüllt")
    else:
        print("2. Bedingung nicht erfüllt")
        
elif ort == "Potsdam":
    print("Alternative:", ort)
    
else:
    print("Keine Bedingung erfüllt")
    pass # Platzhalter, falls hier später Logik eingefügt werden soll
```

### Bedingte Zuweisung (Ternary Operator)
Für einfache Entscheidungen innerhalb einer einzigen Zeile gibt es in Python den sogenannten ternären Operator. 
Die Syntax lautet: `[Wert_wenn_wahr] if [Bedingung] else [Wert_wenn_falsch]`.

Dies lässt sich auch verschachteln, um mehrere Bedingungen auf einmal zu prüfen, was den Code bei simplen Zuweisungen deutlich verkürzt.

```python
a = 10
b = 20

# Einfacher Ternary Operator
kleinere_zahl = a if a < b else b
print(kleinere_zahl, "ist die kleinere Zahl")

print("a ist größer" if a > b else "b ist größer")

# Verschachtelter Ternary Operator
n = -5
ergebnis = "positiv" if n > 0 else "negativ" if n < 0 else "Null"
print("Die Zahl n ist:", ergebnis)
```

### Structural Pattern Matching (`match` / `case`)
Seit Python 3.10 gibt es das sogenannte "Structural Pattern Matching", das ähnlich wie ein `switch/case`-Konstrukt in anderen Programmiersprachen funktioniert. 

Es prüft einen Wert gegen verschiedene Muster (Patterns). Der Unterstrich `_` fungiert hierbei als Wildcard und fängt alle Fälle ab, die nicht durch vorherige `case`-Anweisungen definiert wurden (entspricht einem `default`-Verhalten).

```python
wert = 10

match wert:
    case 5:
        print("Wert ist 5")
    case 10:
        print("Wert ist 10")
    case _:
        print("Wert ist weder 5, noch 10")
```
