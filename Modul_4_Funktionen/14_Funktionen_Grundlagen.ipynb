## 14 - Funktionen (Grundlagen, Parameter & Return)

Funktionen sind zusammengefasste Codeblöcke, die dazu dienen, Redundanzen zu vermeiden (DRY-Prinzip: "Don't Repeat Yourself"). Anstatt identische Logik mehrfach zu schreiben, wird diese einmal in einer Funktion definiert und kann anschließend beliebig oft aufgerufen werden.

Die allgemeine Syntax zur Definition einer Funktion in Python lautet `def funktionsname():`.

```python
# Definition einer einfachen Funktion
def zeige_begruessung():
    print("Willkommen im System.")

# Aufruf der Funktion
zeige_begruessung()
```

### Parameter, Rückgabewerte (`return`) und Type Hinting
Funktionen können Eingabewerte (Argumente/Parameter) entgegennehmen und auf deren Basis Operationen ausführen. Mittels des Schlüsselworts `return` kann das Resultat der Operation an den Aufrufer zurückgegeben werden, anstatt es nur auf der Konsole auszugeben.

**Type Hinting**
In der professionellen Softwareentwicklung wird dringend empfohlen, Datentypen für Parameter und Rückgabewerte explizit anzugeben (Type Hinting). Dies hat keine Auswirkungen auf die Laufzeit, ermöglicht aber IDEs eine automatische Fehlererkennung und verbesserte Autovervollständigung.

```python
# Funktion mit Parametern, Type Hinting und Rückgabewert
# Berechnet den Body-Mass-Index (BMI) = Gewicht in kg / (Körpergröße in m)^2
def berechne_bmi(gewicht_kg: float, groesse_m: float) -> float:
    bmi = gewicht_kg / (groesse_m ** 2)
    return round(bmi, 1)

# Die Rückgabe kann in einer Variablen gespeichert und weiterverarbeitet werden
patienten_bmi = berechne_bmi(85.0, 1.85)
print("Berechneter BMI:", patienten_bmi)
```

### Dokumentation: Docstrings
Um Funktionen verständlich zu dokumentieren, werden in Python sogenannte Docstrings (Documentation Strings) verwendet. Diese werden direkt unter der Funktionsdefinition in dreifachen Anführungszeichen `"""` notiert. 

Die integrierte Python-Funktion `help()` kann diesen Docstring auslesen und als Hilfe für andere Entwickler auf der Konsole ausgeben.

```python
def berechne_summe(a: int = 0, b: int = 0) -> int:
    """
    Berechnet die Summe der Werte aus a und b als Ganzzahl.
    
    Parameter:
    a (int): Der erste Summand (Standardwert: 0)
    b (int): Der zweite Summand (Standardwert: 0)
    
    Rückgabewert: 
    int: Die berechnete Summe.
    """
    return int(a + b)

# Auslesen der Dokumentation
help(berechne_summe)
```

### Call-by-Object-Reference (Veränderbarkeit in Funktionen)
Ein kritisches Konzept in Python ist das Verhalten von Datentypen, wenn sie als Parameter in eine Funktion übergeben werden:

1. **Nicht veränderbare Datenstrukturen (Immutable):** Datentypen wie Integer, Float oder Strings ändern sich nicht innerhalb einer Funktion. Bei einer Neuzuweisung wird ein neues Objekt im Arbeitsspeicher erzeugt. Nach Beendigung der Funktion wird das neue Objekt verworfen, solange es nicht zurückgegeben wird. 
2. **Veränderbare Datenstrukturen (Mutable):** Datentypen wie Listen oder Dictionaries werden als Speicherreferenz übergeben. Werden sie innerhalb der Funktion modifiziert (z. B. mit `.append()`), ändert sich das Originalobjekt global.

```python
# 1. Verhalten bei unveränderbaren Datentypen (Integer)
a = 5
print("Speicherort von a:", hex(id(a)))

def aendere_zahl(x):
    print("Speicherort von x (Start):", hex(id(x)))
    x = 3 # Hier wird ein neues Objekt erzeugt
    print("Speicherort von x (Ende):", hex(id(x)))

aendere_zahl(a)
print("Wert von a (unverändert):", a)

print("-" * 30)

# 2. Verhalten bei veränderbaren Datentypen (Liste)
l = ["Hallo", "Welt"]
print("Speicherort von l:", hex(id(l)))

def aendere_liste(x):
    print("Speicherort von x:", hex(id(x)))
    x.append("!!!") # Das Originalobjekt wird manipuliert

aendere_liste(l)
print("Wert von l (wurde verändert):", l)
```

### Mutable Default Arguments (Ein klassischer Bug)
Da veränderbare Datenstrukturen als Referenz behandelt werden, dürfen sie **niemals** als Standardwert (Default-Parameter) in einer Funktionsdefinition verwendet werden (z. B. `def funktion(liste = []):`). 

Python wertet Standard-Parameter nur ein einziges Mal bei der Definition der Funktion aus. Wird die Funktion mehrfach aufgerufen, teilen sich alle Aufrufe dieselbe Liste im Arbeitsspeicher, was zu unvorhersehbarem Daten-Leakage führt.

**Best Practice:** Es wird standardmäßig `None` übergeben und die Liste erst innerhalb der Funktion erzeugt.

```python
# FALSCH (Gefährlich):
def falsche_protokollierung(eintrag, protokoll=[]):
    protokoll.append(eintrag)
    return protokoll

print("Falsch 1:", falsche_protokollierung("Start"))
print("Falsch 2:", falsche_protokollierung("Stopp")) # "Start" ist fälschlicherweise noch enthalten!

print("-" * 30)

# RICHTIG (Best Practice):
def richtige_protokollierung(eintrag, protokoll=None):
    if protokoll is None:
        protokoll = []
    protokoll.append(eintrag)
    return protokoll

print("Richtig 1:", richtige_protokollierung("Start"))
print("Richtig 2:", richtige_protokollierung("Stopp")) # Saubere, neue Liste
```
