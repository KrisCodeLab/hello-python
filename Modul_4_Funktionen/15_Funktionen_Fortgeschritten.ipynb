Neben Standard-Parametern bietet Python erweiterte Konzepte für die dynamische Parameterübergabe, anonyme Funktionen und speichereffiziente Iterationen.

### Variable Parameter: `*args` und `**kwargs`
Da ein echtes Überladen von Funktionen (Method Overloading) in Python nicht existiert, werden variable Parameter genutzt, um eine unbestimmte Anzahl von Argumenten entgegenzunehmen.
* **`*args` (Arguments):** Sammelt beliebig viele Positionsargumente und speichert sie innerhalb der Funktion als Tupel.
* **`**kwargs` (Keyword Arguments):** Sammelt beliebig viele benannte Argumente (Schlüsselwort-Argumente) und speichert sie als Dictionary.

```python
# Definition einer Funktion mit variablen Parametern
def zeige_daten(*args, **kwargs):
    print("Positionsargumente (Tupel):", args)
    print("Schlüsselwortargumente (Dict):", kwargs)

# Aufruf mit einer beliebigen Anzahl von Parametern
zeige_daten("Test", 42, 3.14, name="Max", beruf="Informatiker")

print("-" * 30)

# Praktischer Anwendungsfall
def konfiguriere_sensor(sensor_id, *messwerte, **einstellungen):
    print(f"Konfiguriere Sensor: {sensor_id}")
    
    # *messwerte ist ein Tupel
    if messwerte:
        durchschnitt = sum(messwerte) / len(messwerte)
        print(f"Durchschnitt der übergebenen Kalibrierungswerte: {durchschnitt}")
        
    # **einstellungen ist ein Dictionary
    for schluessel, wert in einstellungen.items():
        print(f"Setze {schluessel} auf {wert}")

# Aufruf mit variabler Parameteranzahl
konfiguriere_sensor("Bodenfeuchte_01", 45, 48, 42, 50, intervall_sekunden=60, aktiv=True)
```

### Anonyme Funktionen (`lambda`)
Eine Lambda-Funktion ist eine kompakte, anonyme Funktion, die in einer einzigen Zeile notiert wird. Sie erfordert keinen Namen (`def`) und kein explizites `return`-Schlüsselwort, da das Ergebnis des Ausdrucks automatisch zurückgegeben wird. 

Syntax: `lambda Parameter: Ausdruck`

In der Praxis werden Lambda-Funktionen häufig als kurzes Argument an andere Funktionen (wie Sortierfunktionen) übergeben, um nicht extra eine eigene Funktion definieren zu müssen.

```python
tupel_liste = [(10, 2), (4, 1), (0, 17)]

# Sortierung mit einer Lambda-Funktion als Sortierschlüssel
# Der Schlüssel ist hier die Summe der beiden Elemente des Tupels
tupel_liste.sort(key=lambda t: t[0] + t[1])

print("Sortiert nach Summe:", tupel_liste)

print("-" * 30)

# Eine komplexe Datenstruktur (Liste aus Dictionaries)
sortiment = [
    {"artikel": "Kaffee", "preis": 8.00},
    {"artikel": "Milch", "preis": 1.00},
    {"artikel": "Butter", "preis": 2.50}
]

# Ohne Lambda müsste man extra eine Funktion schreiben:
# def sortier_schluessel(element):
#     return element["preis"]

# sortiment.sort(key=sortier_schluessel)

# Mit Lambda: 
# Wir sagen der sort-Methode direkt in einer Zeile, dass sie den Wert bei "preis" als Maßstab nehmen soll.
sortiment.sort(key=lambda element: element["preis"])

for produkt in sortiment:
    print(f"{produkt['artikel']}: {produkt['preis']} EUR")
```

### Generatoren und das Schlüsselwort `yield`
Während `return` eine Funktion vollständig beendet und einen Wert zurückgibt, unterbricht `yield` die Ausführung der Funktion, liefert einen Zwischenwert und speichert den aktuellen Zustand. Beim nächsten Aufruf wird die Ausführung exakt an dieser Stelle fortgesetzt. 

Funktionen, die `yield` nutzen, werden als Generatoren bezeichnet. Generatoren sind extrem speichereffizient, da sie Elemente zur Laufzeit ("on the fly") generieren, anstatt eine komplette Datenstruktur im Arbeitsspeicher aufzubauen.

**Der praktische Nutzen:**
Bei der Verarbeitung von extrem großen Dateien (z. B. Log-Dateien mit Millionen Zeilen) oder Endlos-Datenströmen (Live-Wetterdaten) würde `return liste` den Arbeitsspeicher (RAM) sofort sprengen. Ein Generator mit `yield` holt sich speichereffizient immer nur den exakt nächsten Wert "on the fly".

```python
# Generator-Funktion
def gen_generator():
    for i in range(0, 3):
        yield i

print("Generator iterieren:")
for element in gen_generator():
    print(element)

print("-" * 30)

import time

def generiere_messdaten(anzahl):
    """
    Simuliert das Auslesen einer riesigen Datenmenge.
    Anstatt alle Daten in eine Liste zu packen (was RAM frisst),
    wird jeder Datensatz einzeln per 'yield' weitergereicht.
    """
    for i in range(anzahl):
        # Simuliert eine aufwendige Berechnung oder Datenbankabfrage
        datensatz = {"id": i, "temperatur": 20 + i, "status": "OK"}
        yield datensatz

print("Starte Datenstrom...")
# Der Generator iteriert, ohne den Speicher mit allen 10.000 Elementen gleichzeitig zu belasten
for messung in generiere_messdaten(10000):
    if messung["id"] > 4: # Wir brechen hier frühzeitig ab, um die Konsole nicht zu fluten
        break
    print(f"Verarbeite: {messung}")
```
