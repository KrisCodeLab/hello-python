## 16 - Rekursion

Eine rekursive Funktion ist eine Funktion, die sich während ihrer Ausführung selbst aufruft. Rekursion wird in der Softwareentwicklung häufig eingesetzt, um komplexe, verschachtelte Probleme (wie das Durchsuchen von Dateiverzeichnissen oder Baumstrukturen) in immer kleinere, gleichartige Teilprobleme zu zerlegen.

### Das Rekursionslimit und der Basis-Fall
Jeder Funktionsaufruf belegt Arbeitsspeicher (den sogenannten Call-Stack). Ruft sich eine Funktion unendlich oft selbst auf, würde das System irgendwann abstürzen. Python schützt das Betriebssystem davor durch ein festes Rekursionslimit. 

Aus diesem Grund benötigt **jede** rekursive Funktion zwingend einen **Basis-Fall (Abbruchbedingung)**, der die Rekursion stoppt, bevor das Limit erreicht wird.

```python
import sys

# Abfragen des aktuellen Limits des Python-Interpreters
limit = sys.getrecursionlimit()

print("Das standardmäßige Rekursionslimit liegt bei:", limit, "Aufrufen")

# Das Limit ließe sich theoretisch überschreiben (z.B. sys.setrecursionlimit(1500)),
# was in der Praxis jedoch oft auf Architekturprobleme im Code hindeutet.
```

### Klassisches Beispiel: Fibonacci-Zahlen
Die Fibonacci-Folge ist ein Standardbeispiel für Rekursion. Jede Zahl berechnet sich aus der Summe der beiden vorherigen Zahlen (`Fib(n) = Fib(n-1) + Fib(n-2)`). 

Um eine Endlosschleife zu verhindern, werden die Basis-Fälle für `n=1` und `n=2` fest definiert.

```python
def fib(n: int) -> int:
    """Rekursive Funktion zur Berechnung der n-ten Fibonacci-Zahl."""
    # 1. Basis-Fälle (Abbruchbedingung)
    if n == 1:
        return 0
    elif n == 2:
        return 1
    
    # 2. Rekursiver Aufruf (Selbstaufruf)
    return fib(n - 1) + fib(n - 2)

# Testaufruf
n_wert = 5
print(f"Die Fibonacci-Zahl für n={n_wert} lautet: {fib(n_wert)}")
```

### Praxis-Beispiel: Rekursive Suche in Datenstrukturen
Rekursion eignet sich hervorragend, um Listen oder Ordnerstrukturen schrittweise abzuarbeiten. Bei jedem Selbstaufruf wird das Problem (die zu durchsuchende Liste) um ein Element verkleinert, bis der Basis-Fall eintritt.

```python
def suche_in_liste(wert: str, daten: list, level: int = 0) -> bool:
    """Durchsucht eine Liste rekursiv nach einem bestimmten Wert."""
    print(f"Aufruflevel {level}: Durchsuche {daten}")

    # Basis-Fall 1: Die Liste ist leer, der Wert wurde nicht gefunden
    if len(daten) == 0:
        return False

    # Basis-Fall 2: Das erste Element der aktuellen Liste ist der Treffer
    if daten[0] == wert:
        return True

    # Rekursiver Aufruf: Die Funktion ruft sich selbst auf, übergibt aber
    # nur noch den Rest der Liste ab Index 1 (Slicing)
    return suche_in_liste(wert, daten[1:], level + 1)

hauptstaedte = ["Berlin", "Rom", "Oslo", "Bern"]
gesucht = "Oslo"

if suche_in_liste(gesucht, hauptstaedte):
    print(f"Erfolg: {gesucht} wurde gefunden!")
else:
    print(f"Fehlschlag: {gesucht} ist nicht in der Liste.")
```
